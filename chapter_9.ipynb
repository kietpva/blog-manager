# Lists

## A list is a sequence
A list in Python is a sequence of values. Like strings, lists support many operations that sequences do: indexing, slicing, and iteration. The difference is that lists are mutable, meaning you can change their contents after they are created. Lists can contain elements of any type, including other lists.

Because lists are sequences, you can use:
- `len()` to get the number of elements
- indexing `list[i]` to access a specific element
- slicing `list[start:end]` to get a sub-list
- `for` loops to iterate through values

Additionally, lists support methods like `append`, `insert`, `remove`, and `pop` to modify the list

**Example: creating a list**

````python
numbers = [1, 2, 3, 4, 5]
words = ["apple", "banana", "cherry"]
mixed = [1, "a", True, [2, 3]]
````

**Example: Length and indexing**

````python
print(len(numbers))    # 5
print(words[1])        # "banana"
````

**Example: Slicing**

````python
print(numbers[1:4])   # [2, 3, 4]
````

**Example: Mutable - Changing a list**

````python
numbers[1] = 10
print(numbers)  # [10, 2, 3, 4, 5]
````

**Example List methods**

````python
words.append("date")
words.remove("banana")
first = words.pop(0)

print(first)
````

## Lists are mutable
The phrase “lists are mutable” means that you can change the contents of a list after it has been created. This is different from strings, which are immutable (cannot be changed directly once created).

For example, you can:
- Change the value of an element at a specific index using assignment
- Add elements to the list using methods like `append()` or `insert()`
- Remove elements with methods like `remove()` or `pop()`
- Replace a slice of a list with new values

Because lists are mutable, u do not need to create a new list every time u want to update it, you can modify the same object in place

**Example: Change an element**

````python
nums = [1, 2, 3]
nums[0] = 10
print(nums)  # [10, 2, 3]
````

**Example: Append a new element**

````python
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits) # ["apple", "banana", "cherry"]
````

**Example: Remove an element**

````python
vals = [1, 2, 3, 4]
vals.remove(3)
print(vals)  # [1, 2, 4]
````

**Example: Replace a slice**

````python
nums = [1, 2, 3, 4, 5]
nums[1:3] = [7, 8]
print(nums)  # [1, 7, 8, 4, 5]
````

## List slices
List slicing works just like string slicing: you can extract a part of a list using the slice expression:

```sql
list[start:end]
```
This returns a new list that contains elements from index `start` up to but not including index `end`. You can also omit `start` or `end`:
- `list[:end]`: from the beginning to `end`
- `list[start:]`: from `start` to the end
- `list[:]`: a copy of the whole list

Slicing does not modify the original list, it creates a new one

Because lists are mutable, you can use slice assignment to change parts of the list:

```sql
list[start:end] = new_values
```

This replaces the specified slice with the contents of `new_values`

**Example: simple slice**

````python
numbers = [1, 2, 3, 4, 5]
print(numbers[1:4])  # [2, 3, 4]
````

**Example: Omit start/end**

````python
print(numbers[:3])  # [1, 2, 3]
print(numbers[3:])  # [4, 5]
print(numbers[:])   # [1, 2, 3, 4, 5]
````

**Example: Slice assignment (modify list)**

````python
numbers = [1, 2, 3, 4, 5]
numbers[1:4] = [7, 8, 9]
print(numbers)  # [1, 7, 8, 9, 5]
````

## List operations
This section explains the basic operations you can perform on lists using operators. Lists support the sequence operators just like strings and tuples:
1. Concatenation(`+`)
   - Combines two lists to make a new one
   - Does not modify the original lists.
2. Reptition(`*`)
   - Repeats a list some number of times
   - Also returns a new list
3. Membership(`in`)
   - Tests whether a value appears in the list
   - Returns `True` or `False`

These operations treat lists as sequences, similar to how you would work with strings

**Example: Concatenation(`+`)**

````python
a = [1, 2, 3]
b = [4, 5]
c = a + b
print(c)  # [1, 2, 3, 4, 5]
````

**Example: Repetition(`*`)**

````python
nums = [0, 1]
print(nums * 3)  # [0, 1, 0, 1, 0, 1]
````

**Example: membership test(`in`)**

````python
names = ["Alice", "Bob", "Charlie"]
print("Bob" in names)   # True
print("Dave" in names)  # False
````

## List methods
List in Python are mutable and come with built-in methods that allow u to modify the list contents. These methods perform common operations such as:

**Important list methods:**
- append(x): Add `x` to the end of the list
- insert(i, x): Insert `x` at index `i`
- remove(x): Remove the first occurrence of `x`
- pop([i]): Remove and return item at index `i`(default last)
- sort(): Sort the list in place
- reverse(): Reverse the list in place
- index(x): Return the index of the first occurrence of `x`

Use these methods to change the list in place without creating a new list

### **Example**:
#### **append(x)**

````python
nums = [1, 2, 3]
nums.append(4)
print(nums)  # [1, 2, 3, 4]
````

#### **insert(i, x)**

````python
names = ["Alice", "Bob"]
names.insert(1, "Carol")
print(names)  # ["Alice", "Carol", "Bob"]
````

#### **remove(x)**

````python
vals = [1, 2, 3, 2]
vals.remove(2)
print(vals)  # [1, 3, 2]
````

#### **pop([i])**

````python
nums = [10, 20, 30]
last = nums.pop()
print(last, nums)  # 30 [10, 20]
first = nums.pop(0)
print(first, nums) # 10 [20]
````

#### **sort()**

````python
nums = [3, 1, 2]
nums.sort()
print(nums)  # [1, 2, 3]
````

#### **reverse()**

````python
nums = [1, 2, 3]
nums.reverse()
print(nums)  # [3, 2, 1]
````

#### **index(x)**

````python
letters = ["a", "b", "c"]
print(letters.index("b"))  # 1
````

## Lists and strings
This section explains the relationship between lists and strings in Python. Although they are different types, they have similarities because both are sequences. You can use the same sequence operations—like indexing, slicing, and iteration—on both lists and strings.

However, one key difference is mutability:

- String are immutable (cannot be changed in place)
- Lists are mutable (can be changed)

Because of this, you can use list operations to build up a sequence of characters more flexibly than with strings

For example, if you want to build words one character at a time, you can accumulate the characters in a list and then convert it to a string using the `join()` method

### **Example**
#### **Iterating over a string and list**

````python
for ch in "hello":
    print(ch)

for item in [1, 2, 3]:
    print(item)
````

#### **Strings are immutable**

````python
word = "hello"
# word[0] = "H"  # Error (cannot modify string)
````

#### **Use list to build a string**

````python
chars = []
for ch in "hello":
    chars.append(ch)

word = "".join(chars)
print(word)  # "hello"
````

## Looping through a list
This section explains how to iterate through the elements of a list using a `for` loop. You can use a `for` loop to visit each element in a list in order, just like you use `for` with strings. The loop variable takes on each element of the list in turn.

A common pattern is to process elements one at a time, such as printing them, accumulating a value, or performing some computation.

You can also use a `for` loop with the `range()` function and `len()` to iterate over the indices of a list if you need both the index and the value

### **Example**
#### **Loop through values**

````python
words = ["apple", "banana", "cherry"]

for word in words:
    print(word)
````

#### **Loop with indices**

````python
nums = [10, 20, 30]
for i in range(len(nums)):
    print(i, nums[i])
````

## Sorting lists
This section explains how to sort a list in Python. The simplest way to sort a list of values (like numbers or strings) is to use the `sort()` method provided by lists
- `list.sort()` sorts the list in place, meaning it changes the original list
- By default, `sort()` arranges the elements in ascending order
- For lists of number, that means from smallest to largest
- For lists of strings, that means in alphabetical order

Because `sort()` modifies the list it is called on, it does not return a new list, ut returns `None`

### **Example**
#### **Sort a list of numbers**

````python
nums = [4, 2, 7, 1]
nums.sort()
print(nums)   # [1, 2, 4, 7]
````

#### **Sort a list of strings**

````python
words = ["banana", "apple", "cherry"]
words.sort()
print(words)   # ['apple', 'banana', 'cherry']
````

## Objects and values
This section explains the important distinction between objects and values in Python.
- A value is a piece of data (like the number `42` or the list `[1, 2, 3]`)
- An object is an entity in memory that has an identity, type, and value
- Variables in Python are names that refer to objects

When you assign a list to a variable, you are making the name refer to the same list object, not creating a copy

```python
a = [1, 2, 3]
b = a
```

After this, both `a` and `b` refer to the same object. If you modify the list through `b`, the change is visible through `a` as well because there is only on list object

This section also discusses how mutablity matters:

- Mutable objects (like lists) can be changed in place
- Immutable objects (like numbers and strings) cannot be changed, operations that seem to change them, actually create new objects

Understanding the difference between objects and values helps explain why certain assignments and mutations behave the way they do in Python.

### **Example**
#### **Shared reference for lists**

````python
a = [1, 2, 3]
b = a
b[0] = 5
print(a)  # [5, 2, 3]
````

#### **Immutable example with strings**

````python
s = "hello"
t = s
s = s + "!"
print(t)  # "hello"
````

## Aliasing
“Aliasing” in Python refers to the situation where two or more variable names refer to the same object. When you assign a list to another variable using simple assignment, you do not make a copy of the list — both variables point to the same underlying list.

As a result, modifying the list through one name affects the other. This can lead to unexpected behavior if you intended to make an independent copy but mistakenly created an alias instead.

To avoid aliasing, you must explicitly create a copy of the list, for example using slicing (`my_list[:]`) or the `list()` constructor.

### **Example**
#### **Aliasing example**

````python
a = [1, 2, 3]
b = a
b[0] = 10
print(a)   # [10, 2, 3]
````

#### **Making a copy to avoid aliasing**

````python
a = [1, 2, 3]
b = a[:]      # copy list
b[0] = 10
print(a)      # still [1, 2, 3]
````
