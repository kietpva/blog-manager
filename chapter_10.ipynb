# Dictionaries

## A dictionary is a mapping

A dictionary is like a list, but more general. In a list, the indices have to be integers; in a dictionary they can be (almost) any type. This means:
- A dictionary contains key-value pairs
- Each key is associated with a value
- This association is called a mpping from keys to values
- You can use a dictionary to look up a value by its key

````python
numbers = {}
numbers['zero'] = 0
numbers['one'] = 1
numbers['two'] = 2

print(numbers)
````

## Creating dictionaries
This section explains how to create dictionaries in Python. A dictionary is a collection of key–value pairs, where each key maps to a corresponding value. You can think of a dictionary as a simple lookup table.

There are several ways to create a dictionary:
1. Using curly braces `{}` with key:value pairs:
   ```python
   d = {'a': 1, 'b': 2, 'c': 3}
   ```
2. Starting with an empty dictionary and adding entries later:
   ```python
   d = {}
   d['a'] = 1
   d['b'] = 2
   ```

### **Example:**
#### **Create a dictionary with literals**

````python
counts = {'alice': 1, 'bob': 2, 'carol': 3}
print(counts)
````

#### **Start with empty and add entries**

````python
d = {}
d['age'] = 20
d['name'] = 'Alice'
print(d)
````

#### **Look up a value**

````python
print(d['name'])   # Alice
````

## **The `in` operator**
This section explains how to use the in operator with dictionaries in Python. When used with a dictionary, the in operator checks whether a given key exists in the dictionary. It does not check the values, only the keys.

```python
if key in d:
    ...
```
This is useful to avoid errors like KeyError when accessing a key that might not be present in the dictionary.

For example, before accessing d[key], you can check:
```python
if key in d:
    value = d[key]
```
This helps make your code safer and prevents crashes due to missing keys.

### **Example**
#### **Key in dictionary**

````python
d = {'name': 'Alice', 'age': 30}

print('name' in d)   # True
print('Alice' in d)  # False (because 'Alice' is a value, not a key)
````

#### **Using `in` to avoid KeyError**

````python
if 'score' in d:
    print(d['score'])
else:
    print("No score found")
````

## A collection of counters
This section introduces the idea of using a dictionary as a collection of counters—that is, a way to count how many times each distinct item appears in a sequence (such as words in a text). Instead of using many separate variables, the dictionary can map each item to its own count.

The basic pattern for counting with a dictionary is:

1. Create an empty dictionary
2. For each item in a list or other sequence:
   - Check if the item is already a key in the dictionary
   - If not, add it with a count of 1
   - If it is, increment its count

This pattern builds a "collection of counters" that makes it easy to aggregate counts for many distinct items without knowing them beforehand

### **Example**
#### **The counting pattern**

````python
sequence = "banana"

counts = {}
for item in sequence:
    if item not in counts:
        counts[item] = 1
    else:
        counts[item] += 1

print(counts)
````

#### **One-line pattern**

````python
sequence = "banana"

counts = {}
for item in sequence:
    counts[item] = counts.get(item, 0) + 1

print(counts)
````

## Looping and dictionaries
This section explains how to use loops with dictionaries. Unlike lists or strings, dictionaries are not sequences in a strict sense, so you don’t loop over them by index. Instead, you can loop directly over the keys of the dictionary.

When you write:
```python
for key in d:
    ...
```

Python iterates over the keys of the dictionary `d`. You can then use each key to access the corresponding value using `d[key]`.
If you want both the key and the value, a useful pattern is to use the `items()` method, which returns pairs of (key, value) that can be unpacked in the loop:

```python
for key, value in d.items():
    ...
```

This allows you to process every entry in the dictionary in a clean and readable way

### **Example**
#### **Loop over keys**

````python
counts = {'a': 3, 'b': 5, 'c': 2}

for key in counts:
    print(key, counts[key])
````

#### **Loop using items()**

````python
counts = {'a': 3, 'b': 5, 'c': 2}

for key, value in counts.items():
    print(key, value)
````

## Lists and dictionaries
This section explains how lists and dictionaries can work together. A list can contain dictionaries and a dictionary can contain lists. This flexibility allows you to represent more complex data structures

A common use case is when a word count program produces a dictionary that maps words to their frequencies. Sometimes you may want to sort the result or process it in ways that benefit from using a list of key–value pairs

For example, if you want to find the most frequent words, you might convert the dictionary into a list of `(value, key)` tuples and then sort that list.

The section also shows how you can build a list of dictionaries to model structured data — for example, a list of records where each record is a dictionary with named fields

### **Example**
#### **Dictionary to list for sorting**

````python
counts = {'a': 3, 'b': 5, 'c': 2}

items = []
for word, count in counts.items():
    items.append((count, word))

items.sort(reverse=True)

print(items)
````

#### **List of dictionaries**

````python
records = []
records.append({'name': 'Alice', 'age': 30})
records.append({'name': 'Bob', 'age': 25})
print(records)
````

## Accumulating a list
This section explains how to build up a list while processing data using a loop, especially in combination with a dictionary. The idea is to accumulate a list of items (e.g., `(value, key)` pairs) so that you can sort or further process the results.

A common pattern is:

1. Start with an empty list
2. Loop through items in a dictionary
3. Append new items(often tuples) to the list in each iteration
4. after the loop, use the list for sorting or other operations

Using an accumalated list of tuples is useful when you want to, for example, sort dictionary items by value (e.g., word frequencies) instead of by key. The dictionary itself cannot be sorted directly because it is not a sequence, converting it to a list of tuples allws you to sort in any desired order

### **Example**
#### **Sort word frequencies**

````python
counts = {'a': 5, 'the': 10, 'and': 7}

items = []
for word, count in counts.items():
    items.append((count, word))

items.sort(reverse=True)

for count, word in items[:10]:
    print(word, count)
````

## Memos
This section explains a simple way to use a dictionary as a memo (short for memorandum) to remember values you have already calculated. A memo helps make a function more efficient by storing results of previous computations so that if the same computation is needed again, you can look up the answer instead of recalculating it.

The basic idea:
- Use a dictionary where the keys represent the inputs to a function, and the values represent the corresponsing outputs
- When the function is called:
  1. First check if the input is already in the memo
  2. If it is, return the stored result
  3. If not, perform the calculation, store it in the memo, and then return the result

Memos are especially useful in recursive functions, where the same subproblems can be computed multiple times

### **Example**
#### **Pattern**

````python
memo = {}

def f(x):
    if x in memo:
        return memo[x]
    # compute result
    result = ...  
    memo[x] = result
    return result
````

#### **Memoizing Factorial**

````python
memo = {}

def factorial(n):
    if n == 0:
        return 1
    if n in memo:
        return memo[n]
    result = n * factorial(n - 1)
    memo[n] = result
    return result
````
