# **Python Extras**
## **Sets**
A `set` is a collection of unique elements, duplicates are automatically removed.

```python
s = set([1,2,2,3])
print(s)   # {1, 2, 3}
```

## **Counters**
A `Counter` counts occurrences of elements like a mutiset 

```python
from collections import Counter
c = Counter("banana")
print(c)  # Counter({'a': 3, 'n': 2, 'b': 1})
```

## **defaultdict**
A `deaultdict` automatically creates default values for missing keys

```python
from collections import defaultdict
d = defaultdict(list)
d["x"].append(10)
print(d)  # defaultdict(<class 'list'>, {'x': [10]})
```

## **Conditional expressions**
A concise syntax for simple if-else expressions

```python
x = 2
y = x if x > 0 else 0
```
