# **Tuples**
## **Tuples are like lists**
In Python, a tuple is a built-in sequence type that is very similar to a list. Like lists:

- Tuples contain a sequence of value
- The values can be of any type
- You can use indexing and slicing to access elements
- You can use sequence operators such as `+` (concatenation) and `*` (repeat)
- Functions like `len()` and `reversed()` work with tuples

However, unlike lists, tules are immutable, which means that once a tuple is created, yoi can't change its elements(more on that in the next section). Because tuples are immutable, they can be used as keys in dictionaries and in other contexts where immutablility is required

### **Example**
#### **Creating a tuple**

```python
t = ('a', 'b', 'c')
print(t)          # ('a', 'b', 'c')
print(type(t))    # <class 'tuple'>
```

#### **Indexing and slicing like lists**

```python
t = ('p','y','t','h','o','n')
print(t[0])       # 'p'
print(t[1:4])     # ('y', 't', 'h')
```

#### **Concatenate and repeat**

```python
t1 = ('a','b')
t2 = ('c','d')
print(t1 + t2)    # ('a','b','c','d')
print(t1 * 3)     # ('a','b','a','b','a','b')
```

#### **Tuple from sequence**

```python
print(tuple('hello'))   # ('h','e','l','l','o')
```
