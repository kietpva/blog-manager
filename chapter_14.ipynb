# **Classes and Functions**
## **Programmer-defined types**
This section introduces the idea of programmer-defined types, which are new kinds of objects you can create in Python to organize both data and behavior. These types are created using the `class` keyword and are also called classes. A programmer-defined type lets you represent real-world or logical concepts (like time, points, rectangles) more naturally than using basic built-in types alone. 
Allen Downey's GitHub Pages

Here’s what happens when you define a class:

1. You write a `class` definition, giving the type a name.
2. The body of the class (usually a docstring at first) describes what the objects will represent.
3. Defining a class creates a class object that acts like a factory for building instances.
4. To create an instance of your new type, you call the class name like a function (e.g., `Time()`).
5. Each instance (object) has its own attributes that can store values. 
Allen Downey's GitHub Pages

Objects created from classes are instances of your custom type, and Python prints their type and memory location unless you define a special formatting method.
### **Example**
#### **Define a new type**

```python
class Time:
    """Represents a time of day."""
```

#### **Create an instance**

```python
lunch = Time()
print(type(lunch))
```

#### **Add an attribute to the object**

```python
lunch.hour = 11
lunch.minute = 59
lunch.second = 1
```

#### **Access attributes**

```python
print(lunch.hour)
```

## **Attributes**
This section explains attributes, which are names associated with data stored inside an object. When you create an instance of a class, you can attach attributes to it to store values specific to that instance.

You can:

- Create attributes by assigning to them with dot notation (e.g., `obj.attribute = value`).
- Access attribute values using dot notation (e.g., `obj.attribute`).
- Use attributes to keep track of state.

Attributes are specific to each instance (unless defined at the class level). They allow objects to hold data, enabling you to represent concepts with multiple pieces of related information (e.g., hours, minutes, and seconds for a `Time` object)

### **Example**
#### **Creating attributes**

```python
class Time:
    """Represents time of day."""

# Create instance
t = Time()

# Add attributes
t.hour = 11
t.minute = 59
t.second = 1
```

#### **Accessing attributes**

```python
print(t.hour)     # 11
print(t.minute)   # 59
print(t.second)   # 1
```

#### **Modify attribute**

```python
t.hour = 4
print(t.hour)     # 4
```

## **Objects as return values**
Functions can create and return objects

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time

t = make_time(10, 45, 20)
print(t.hour)     # 10
```

## **Object mutability**
If you modify an object inside a function, the original object changes

```python
def increment(time, seconds):
    time.second += seconds

t = make_time(10, 10, 1)
print(t.second)
increment(t, 30)
print(t.second)   #31
```
