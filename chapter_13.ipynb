# **Files and Databases**
## **Filenames and paths**
This section explains how Python handles file names and file paths when working with the file system.

A filename is the name of a file, such as `"words.txt"`. When you open a file with just a filename, Python looks in the current directory (the folder where the program is running). You can also specify a path, which tells Python exactly where to find the file.

Types of paths:

Relative path: describes the location relative to the current directory.

Example: `"data/words.txt"` refers to a subfolder `data` under the current directory.

Absolute path: specifies the full path from the root of the file system.

Example (Windows): `"C:\\Users\\name\\data\\words.txt"`

Example (Linux/Mac): `"/home/name/data/words.txt"`

Python supports using both types of paths. On Windows, backslashes `\` are used, but in string literals you often need to escape them (`"\\"`) or use raw strings (`r"C:\path"`).

To work with paths in a platform-independent way, Python provides the `os.path` module (and in newer code the `pathlib` module), which make it easier to build and manipulate file paths correctly.

### **Example**
#### **Open file using relative path**

````python
fin = open("words.txt")
````

#### **Open file with subfolder path**

````python
fin = open("data/words.txt")
````

#### **Windows absolute path**

````python
fin = open("D:\\python-project\\python-training\\words.txt")

# or

fin = open(r"D:\python-project\python-training\words.txt")
````

## **f-strings**
This section introduces f-strings, a modern and convenient way to format strings in Python. An f-string starts with the letter `f` before the opening quote and allows you to embed expressions directly inside the string using curly braces `{}`.

Instead of using older formatting methods like concatenation (`+`) or `.format()`, f-strings provide a clear, readable way to combine text with variables or expressions.

For example:
```python
name = "Alice"
age = 30
print(f"{name} is {age} years old")
```

The expressions inside {} are evaluated and converted to strings automatically.

F-strings can also include formatting specifiers such as width, decimal places, or alignment.
### **Example**
#### **Simple f-string**

````python
name = "Python"
version = 3.14
print(f"I love {name} {version}!")
````

#### **Expressions inside f-string**

````python
length = 5
width = 3
print(f"Area = {length * width}")
````

#### **Using format specifiers**

````python
pi = 3.14159265
print(f"Pi is approximately {pi:.2f}")
````
