# Classes and Objects
This chapter uses geometric objects (Point, Linem Rectangle) to deepen your understanding of object-oriented programming

It introduces:
- Object composition
- Identity vs equivalence
- Shallow vs deep copy
- Polymorphism
- Debugging mutable objects

## Creating a Point
A `Point` represents a location in 2D space. It stores `x` and `y` as attributes and uses `__init__` to initialize data.

```python
class Point:
    """Represents a point in 2D space"""

    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p = Point(3, 4)
print(p)
```

## **Creating a Line**
A `Line` object contains two `Point` objects as endpoints

```python
class Line:
    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2

    def __str__(self):
        return f"{self.p1} -> {self.p2}"

a = Point(0, 0)
b = Point(3, 4)
line = Line(a, b)
print(line)
```

## **Equivalence and Identity**

- `is` -> same object
- `==` -> same value (only if `__eq__` is defined)

```python
p1 = Point(1, 2)
p2 = Point(1, 2)

p1 == p2   # False
p1 is p2   # False
```

**Override `__eq__`**

```python
p1 == p2   # True
```

## **Creating a Rectangle**
A `Retangle` has:
- width
- height
- a corner `Point`

```python
class Rectangle:
    def __init__(self, width, height, corner):
        self.width = width
        self.height = height
        self.corner = corner

    def __str__(self):
        return f"Rect({self.width}x{self.height}) at {self.corner}"

rect = Rectangle(10, 5, Point(0, 0))
print(rect)
```

## **Changing Rectangles**
Mutating shared objects can cause unexpected behavior

```python
r1 = Rectangle(10, 5, Point(0, 0))
r2 = r1

r2.corner.x = 100
print(r1.corner.x)  # 100
```
