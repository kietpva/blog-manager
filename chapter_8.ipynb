# Strings and Regular Expressions

## A string is a sequence
In Python, a string is a sequence of characters. This means you can treat it like other sequences (such as lists or tuples). You can access individual characters using indexing with square brackets (`[]`), and you can iterate over a string with a `for` loop.

Because strings are sequences, you can apply many sequence operations to them, such as:
- Indexing (`word[0]`, `word[1]`, …)
- Slicing (`word[1:4]`)
- Using `len()` to find the length
- Looping with `for`

However, strings are immutable, meaning once created they cannot be changed. If you want to “modify” a string, you must create a new string.

**Example: Accessing characters**

````python
word = "hello"
print(word[0])   # h
print(word[1])   # e
````

**Example: Slicing**

````python
word = "hello"
print(word[1:4])   # ell
````

**Example: Iteration**

````python
for ch in "world":
    print(ch)
````

**Example: Immutability**

````python
word = "hello"
# This will cause an error:
# word[0] = 'H'
new_word = 'H' + word[1:]
print(new_word)  # Hello
````

## String slices
String slicing is a way to extract a substring from a string. A slice uses the syntax:

```sql
s[start:end]
```

It returns a new string containing the characters from index `start` up to but not including index `end`.

Slices also support optional values:
- `s[:end]` means from the beginning (index 0) up to `end`.
- `s[start:]` means from `start` to the end.
- `s[:]` means the whole string.
- Negative indices count from the end (`-1` is the last character).

Slicing creates a new string and does not modify the original. It can be used to build more complex string operations.

**Example: Basic slicing**

````python
word = "banana"
print(word[1:4])   # "ana"
````

**Example: Omitted start**

````python
word = "banana"
print(word[:3])    # "ban"
````

**Example: Omitted end**

````python
word = "banana"
print(word[3:])    # "ana"
````

**Example: Negative indices**

````python
word = "banana"
print(word[-3:-1])  # "na"
````

**Example: Whole string slice**

````python
word = "think"
print(word[:])   # "think"
````

## Strings are immutable
This section explains that strings in Python are immutable, which means once a string is created, you cannot change its characters in place. Attempts to modify a character directly using indexing will result in an error.

Because strings cannot be modified, if you want a “changed” version of a string, you must create a new string. For example, instead of trying to assign to `s[0]`, you build a new string using concatenation and slicing.

This immutability is important to understand because it affects how string operations behave and why some coding patterns (like concatenating many substrings in a loop) can be less efficient.

**Example:**

❌ Trying to modify a character

````python
word = "hello"
word[0] = "H"   # ❌ This will cause an error
````

✔️ Correct way: create a new string

````python
word = "hello"
new_word = "H" + word[1:]
print(new_word)  # "Hello"
````

## String comparison
This section explains how string comparison works in Python. Strings can be compared using the standard comparison operators like `==`, `!=`, `<`, `>`, `<=`, and `>=`. When you compare two strings, Python compares them lexicographically, meaning it compares each character in order using their ASCII/Unicode values.

A key point is that string comparison is case-sensitive (e.g., `'Zebra'` < `'apple'` may be `True` because uppercase letters have smaller ASCII/Unicode values than lowercase). Also, shorter strings can be less than longer strings if they match up to the shorter length.

You can use string comparison to sort lists of strings or to check ordering.

**Example: Simple equality**

````python
print("apple" == "apple")   # True
print("apple" == "Apple")   # False
````

**Example: Lexicographical order**

````python
print("banana" < "cherry")   # True
print("apple" < "banana")    # True
print("Zebra" < "apple")     # True or False depending on encoding
````

## String methods
Strings in Python come with built-in methods—functions that are bound to string objects and help you perform common tasks without writing manual loops. Method calls use the syntax:

```python
object.method(arguments)
```

For strings, methods can:
- Transform case (`lower()`, `upper()`, `capitalize()`)
- Search and find substrings (`find()`, `startswith()`, `endswith()`)
- Modify white space (`strip()`, `lstrip()`, `rstrip()`)
- Replace parts of the string (`replace()`)
- Split or join text (`split()`, `join()`)

Unlike operators like `in` or slicing, methods encapsulate specific behavior, making code cearer and more compact

**Example: Case conversion**

````python
word = 'Hello World'
print(word.lower())     # 'hello world'
print(word.upper())     # 'HELLO WORLD'
print(word.capitalize())# 'Hello world'
````

**Example: Finding substrings**

````python
word = 'banana'
print(word.find('n'))       # 2
print(word.startswith('ba'))# True
print(word.endswith('na'))  # True
````

**Example: Strip whitespace**

````python
text = '   hello   '
print(text.strip())   # 'hello'
print(text.lstrip())  # 'hello   '
print(text.rstrip())  # '   hello'
````

**Example: Replace substrings**

````python
word = 'banana'
print(word.replace('na', 'XY'))  # "baXYXY"
print(word.replace('ba', 'XY'))  # "XYnana"
````

**Example: Split and join**

````python
sentence = 'one two three'
parts = sentence.split()       # ['one','two','three']
joined = '-'.join(parts)       # 'one-two-three'
print(parts)
print(joined)
````

## Writing files
This section explains how to write text to a file in Python. To write to a file, you first open it in a mode that allows writing. The built-in `open()` function is used with a second argument specifying the mode:
- `"w"`: write mode (creates a new file or truncates an existing file)
- `"a"`: append mode (adds to the end of an existing file)

Once the file is open, you can use the write() method to write strings to it. Because `write()` only writes raw text, only writes raw text, you often need to add your own newline characters(`\n`) if you want each string on a separate line

It is good practice to close the file after writing to ensure all data is flushed and saved

**Example: Basic file writing**

````python
f = open('output.txt', 'w')
f.write("Hello\n")
f.write("World\n")
f.close()
````

**Example: Appending mode**

````python
f = open('output.txt', 'a')
f.write("Another line\n")
f.close()
````

## Find and replace
This section introduces two common string operations: finding a substring and replacing parts of a string.

### Finding
- The `find()` method returns the index of the first occurrence of a substring
- If the substring is not found, `find()` returns `-1`

### Replacing
- The `replace()` method returns a new string where all occurrences of a substring are replaced with another string
- Since strings are immutable, `replace()` does not change the original string — it returns a modified copy

These methods are useful for analyzing and manipulating strings without writing manual loops

**Example: Finding a substring**

````python
word = "banana"
print(word.find("a"))   # 1
print(word.find("na"))  # 2
print(word.find("x"))   # -1(not found)
````

**Example: Replacing substrings**

````python
word = "banana"
new_word = word.replace("na", "XY")
print(new_word)  # "baXYXY"
````
