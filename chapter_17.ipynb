# **Inheritance**
## **Representing Cards**
The chapter begins by showing how to represent playing cards in Python using a class with `suit` and `rank` attributes.

````python
class Card:
    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, 'Ace', '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return f"{Card.rank_names[self.rank]} of {Card.suit_names[self.suit]}"
    
    def __eq__(self, other):
        return self.suit == other.suit and self.rank == other.rank
    
    def __lt__(self, other):
        return (self.suit, self.rank) < (other.suit, other.rank)

queen = Card(1, 12)
print(queen)   # Queen of Diamonds
````

## **Card Attributes**
- A `Card` object has instance attributes: `suit`, `rank`
- It also has class variables: `suit_name`, `rank_names`
- Class variables are shared by all instances

```python

    def __init__(self, suit, rank):

        self.rank = rank    # instance attribute
```

Define a total ordering so cards can be sorted
```python
    def __lt__(self, other):
        return (self.suit, self.rank) < (other.suit, other.rank)
```

````python
cards = [
    Card(3, 13),  # King of Spades
    Card(0, 1),   # Ace of Clubs
    Card(1, 12),  # Queen of Diamonds
]

cards.sort()

for c in cards:
    print(c)
````

**Trick**

- `lt` -> less than -> `<`
- `gt` -> greater than -> `>`
- `le` -> less equal -> `<=`
- `ge` -> greater equal -> `>=`
- `eq` -> equal -> `==`
- `ne` → not equal -> `!=`
