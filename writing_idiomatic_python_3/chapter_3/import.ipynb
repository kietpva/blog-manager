# **Import**
## **Arrange your `import` statements in a standard order**
**❌ DON'T**

```python
from myapp.utils import format_date
import requests
import os
from flask import Flask
from myapp.models import User
import sys
```

**✔️ DO**

```python
# 1. Standard library
import os
import sys

# 2. Third-party libraries
import requests
from flask import Flask

# 3. Local application imports
from myapp.models import User
from myapp.utils import format_date
```

## **Prefer absolute imports to relative imports**
**❌ DON'T**

```python
from .utils import parse
from ..helpers import parse


def run(data):
    return parse(data)
```

**✔️ DO**

```python
import package.sub_package.utils as utils
import package.helpers as helpers


def run(data):
    return utils.parse(helpers.parse(data))
```

## **Do not use from foo import * to import the contents of a module**
**❌ DON'T**

```python
from foo import *
```

**✔️ DO**

```python
from foo import bar, baz, qux, quux, quuux

# or even better...
import foo
```

## **Use the `tuples` to organize a long list of modules to `import`**
**❌ DON'T**

```python
from math import sin
from math import cos
from math import tan
from math import sqrt
from math import pi
from math import e
```

**✔️ DO**
