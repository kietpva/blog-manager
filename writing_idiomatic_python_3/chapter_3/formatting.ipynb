# **Formatting**
## **Use all capital letters when declaring global constant values**
**❌ DON'T**

```python
taxRate = 0.1
MaxRetries = 5
default_timeout = 30
```

**✔️ DO**

```python
TAX_RATE = 0.1
MAX_RETRIES = 5
DEFAULT_TIMEOUT = 30
API_BASE_URL = "https://api.example.com"
```

## **Avoid placing multiple statements on a single line**
**❌ DON'T**

```python
a = 1; b = 2; c = a + b
print(c)
```

**✔️ DO**

```python
a = 1
b = 2
c = a + b
print(c)
```
