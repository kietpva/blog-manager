# **Generators**
## **Prefer a generator expression to a list comprehension for simple iteration**
**❌ DON'T**

```python
users = ["alice", "bob", "carol"]

upper_names = [name.upper() for name in users]

for name in upper_names:
    print(name)
```

**✔️ DO**

```python
users = ["alice", "bob", "carol"]

for name in (name.upper() for name in users):
    print(name)
```

## **Use a generator to lazily load infinite sequencesv**
**❌ DON'T**

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

p = Product("Book", 100)
print(p.price)
```

**✔️ DO**

```python
class Product:
    TAX_RATE = 0.1

    def __init__(self, name, base_price):
        self.name = name
        self._base_price = base_price

    @property
    def price(self):
        return self._base_price * (1 + self.TAX_RATE)

p = Product("Book", 100)
print(p.price)
```
