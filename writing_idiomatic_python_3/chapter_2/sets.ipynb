# **Sets**
## **Understand and use the mathematical `set` operations**
**❌ DON'T**

```python
def get_list_of_most_popular_users():
    return ["alice", "bob", "carol"]

def get_list_of_most_active_users():
    return ["bob", "dave", "carol"]

def get_both_popular_and_active_users():
    # Assume the following two functions each return a
    # list of user names
    most_popular_users = get_list_of_most_popular_users()
    most_active_users = get_list_of_most_active_users()

    popular_and_active_users = []
    for user in most_active_users:
        if user in most_popular_users:
            popular_and_active_users.append(user)

    return popular_and_active_users

get_both_popular_and_active_users()
```

**✔️ DO**

```python
def get_both_popular_and_active_users():
    # Assume the following two functions each return a
    # list of user names
    return set(get_list_of_most_active_users()) & set(get_list_of_most_popular_users())

get_both_popular_and_active_users()
```

# **Use a `set comprehension` to generate sets concisely**
**❌ DON'T**

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

squares = set()
for n in numbers:
    squares.add(n * n)

print(squares)
```

**✔️ DO**

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

squares = {n * n for n in numbers}

print(squares)
```

## **Use sets to determine if two lists share any common values**
**❌ DON'T**

```python
list_a = [1, 2, 3, 4]
list_b = [5, 6, 3, 8]

found = False
for x in list_a:
    if x in list_b:
        found = True
        break

print(found)
```

**✔️ DO**

```python
list_a = [1, 2, 3, 4]
list_b = [5, 6, 3, 8]

common = set(list_a) & set(list_b)

print(bool(common))
```
