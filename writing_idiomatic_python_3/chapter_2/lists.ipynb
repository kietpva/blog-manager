# **Lists**
## **Use a `list comprehension` to create a transformed version of an existing list**
**❌ DON'T**

```python
numbers = [1, 2, 3, 4, 5, 6]
result = []

for n in numbers:
    if n % 2 == 0:
        result.append(n * n)

print(result)
```

**✔️ DO**

```python
numbers = [1, 2, 3, 4, 5, 6]

result = [n * n for n in numbers if n % 2 == 0]

print(result)
```
