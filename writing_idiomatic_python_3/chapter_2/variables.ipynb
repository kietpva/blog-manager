# **Variables**
## **Use multiple assignment to condense variables all set to the same value**
**❌ DON'T**

```python
x = "foo"
y = "foo"
z = "foo"
```

**✔️ DO**

```python
x = y = z = "foo"
```

## **Avoid using a temporary variable when performing a `swap` of two values**
**❌ DON'T**

```python
foo = "Foo"
bar = "Bar"

temp = foo
foo = bar
bar = temp

print(foo, bar, temp)
```

**✔️ DO**

```python
foo = "Foo"
bar = "Bar"
(foo, bar) = (bar, foo)

print(foo, bar, temp)
```
