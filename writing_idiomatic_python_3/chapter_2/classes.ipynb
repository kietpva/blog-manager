# **Classes**
## **Always use `self` or a `@classmethod` when referring to a class’s attributes**
**❌ DON'T**

```python
class Blog:
    __tablename__ = "blog"

    def table_name(self):
        return Blog.__tablename__


class TechBlog(Blog):
    __tablename__ = "tech_blog"


blog = Blog()
tech_blog = TechBlog()

print(blog.table_name())
print(tech_blog.table_name())
```

**✔️ DO**

```python
class Blog:
    __tablename__ = "blog"

    @classmethod
    def table_name(cls):
        return cls.__tablename__


class TechBlog(Blog):
    __tablename__ = "tech_blog"


print(Blog.table_name())
print(TechBlog.table_name())
```

## **Don't use classes solely as a form of encapsulation**
**❌ DON'T**

```python
class MathUtils:
    def add(a, b):
        return a + b

    def subtract(a, b):
        return a - b

print(MathUtils.add(3, 2))
print(MathUtils.subtract(5, 1))
```

**✔️ DO**

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

print(add(3, 2))
print(subtract(5, 1))
```
