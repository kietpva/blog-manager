# **Classes**
## **Always use `self` or a `@classmethod` when referring to a class’s attributes**
**❌ DON'T**

```python
class Blog:
    __tablename__ = "blog"

    def table_name(self):
        return Blog.__tablename__


class TechBlog(Blog):
    __tablename__ = "tech_blog"


blog = Blog()
tech_blog = TechBlog()

print(blog.table_name())
print(tech_blog.table_name())
```

**✔️ DO**

```python
class Blog:
    __tablename__ = "blog"

    @classmethod
    def table_name(cls):
        return cls.__tablename__


class TechBlog(Blog):
    __tablename__ = "tech_blog"


print(Blog.table_name())
print(TechBlog.table_name())
```

## **Don't use classes solely as a form of encapsulation**
**❌ DON'T**

```python
class MathUtils:
    def add(a, b):
        return a + b

    def subtract(a, b):
        return a - b

print(MathUtils.add(3, 2))
print(MathUtils.subtract(5, 1))
```

**✔️ DO**

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

print(add(3, 2))
print(subtract(5, 1))
```

## **Use the isinstance function to determine the type of an object**
**❌ DON'T**

```python
def process(value):
    if type(value) == int:
        print("int:", value * 2)

process(10)
```

**✔️ DO**

```python
def process(value):
    if isinstance(value, int):
        print("int:", value * 2)

process(10)

class MyInt(int):
    pass

process(MyInt(10))
```

## **Use leading underscores in function and variable names to denote "private" data**
**❌ DON'T**

```python
class User:
    def __init__(self, password):
        self.password = password

    def hash_password(self):
        return f"hashed-{self.password}"

user = User("123456")
print(user.password)
```

**✔️ DO**

```python
class User:
    def __init__(self, password):
        self._password = password  # internal

    def _hash_password(self):
        return f"hashed-{self._password}"

user = User("123456")
print(user._password)
```

## **Use properties to "future-proof" your class implementation**
**❌ DON'T**

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, value):
        self._price = value


p = Product("Book", 100)
print(p.get_price())
```

**✔️ DO**

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price

    @property
    def price(self):
        return self._price

p = Product("Book", 100)
print(p.price)
```

## **Use `__repr__` for a machine-readable representation of a class**
**❌ DON'T**

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

u = User("Alice", 30)
print(u)
print([u])
```

**✔️ DO**

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"User(name={self.name!r}, age={self.age!r})"

u = User("Alice", 30)
print(u)
print([u])
```
