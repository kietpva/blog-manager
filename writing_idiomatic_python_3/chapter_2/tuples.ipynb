# **Tuples**
## **Use collections.namedtuple to make tuple-heavy code more clear**
**❌ DON'T**

```python
employee = ("Kiet", 27, "Developer")

name = employee[0]
age = employee[1]
role = employee[2]

print(name, age, role)
```

**✔️ DO**

```python
from collections import namedtuple

Employee = namedtuple("Employee", ["name", "age", "role"])

employee = Employee(age=27, role="Developer", name="Kiet",)

print(employee.name)
```
