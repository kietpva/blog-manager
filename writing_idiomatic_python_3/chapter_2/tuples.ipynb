# **Tuples**
## **Use collections.namedtuple to make tuple-heavy code more clear**
**❌ DON'T**

```python
employee = ("Kiet", 27, "Developer")

name = employee[0]
age = employee[1]
role = employee[2]

print(name, age, role)
```

**✔️ DO**

```python
from collections import namedtuple

Employee = namedtuple("Employee", ["name", "age", "role"])

employee = Employee(age=27, role="Developer", name="Kiet",)

print(employee.name)
```

## **Use _ as a placeholder for data in a tuple that should be ignored**
**❌ DON'T**

```python
user = ("Kiet", 25, "Developer")

name, age, temp1 = user
print(name, age)
```

**✔️ DO**

```python
user = ("Kiet", 25, "Developer")

name, age, _ = user
print(name, age,)
```

## **Use `tuples` to unpack data**
**❌ DON'T**

```python
point = (3, 5)

x = point[0]
y = point[1]

print(x, y)
```

**✔️ DO**

```python
point = (3, 5)

x, y = point
print(x, y)
```

## **Use a tuple to return multiple values from a function**
**❌ DON'T**


```python
def divide(a, b):
    result = {}
    result["quotient"] = a // b
    result["remainder"] = a % b
    return result


r = divide(10, 3)
print(r["quotient"], r["remainder"])
```

**✔️ DO**

```python
def divide(a, b):
    return a // b, a % b


q, r = divide(10, 3)
print(q, r)
```
