## **Use the `enumerate` function in loops instead of creating an “index” variable**

**❌ DON'T**

```python
names = ['Alice', 'Bob', 'Charlie']

i = 0
for name in names:
    print(i, name)
    i += 1
```

**✔️ DO**

```python
names = ['Alice', 'Bob', 'Charlie']

for i, name in enumerate(names):
    print(i, name)
```

## **Use the `in` keyword to iterate over an `iterable`**
**❌ DON'T**

```python
names = ['Alice', 'Bob', 'Charlie']

for i in range(len(names)):
    print(names[i])
```

**✔️ DO**

```python
names = ['Alice', 'Bob', 'Charlie']

for name in names:
    print(name)
```

## **Use `else` to execute code after a `for` loop concludes**
**❌ DON'T**

```python
nums = [1, 3, 5, 7]
found = False

for n in nums:
    if n == 4:
        found = True
        break

if not found:
    print('Not found')
```

**✔️ DO**

```python
nums = [1, 3, 5 ,7]

for n in nums:
    if n == 4:
        print('Found')
        break
else:
    print('Not found')
```
