# **Exceptions**
## **Don’t be afraid to use Exceptions**
**❌ DON'T**

```python
def divide(a, b):
    if b == 0:
        return None
    return a / b

result = divide(10, 0)

if result is None:
    print("Error occurred")
else:
    print(result)
```

**✔️ DO**

```python
def divide(a, b):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("a and b must be numbers")

    if b == 0:
        raise ZeroDivisionError("Division by zero")

    return a / b

try:
    result = divide(10, 0)
    print("Result:", result)
except TypeError as e:
    print("Type error:", e)
except ZeroDivisionError as e:
    print("Math error:", e)
```

## **Use Exceptions to write code in an “EAFP” style**
**❌ DON'T**

```python
def get_first_item(lst):
    if isinstance(lst, list):
        if len(lst) > 0:
            return lst[0]
        else:
            return None
    else:
        return None

print(get_first_item([10, 20, 30]))
print(get_first_item([]))
print(get_first_item("abc"))
print(get_first_item(123))
```

**✔️ DO**

```python
def get_first_item(lst):
    try:
        return lst[0]
    except (TypeError, IndexError):
        return None

print(get_first_item([10, 20, 30]))
print(get_first_item([]))
print(get_first_item("abc"))
print(get_first_item(123))
```

## **Avoid “swallowing” useful exceptions with bare except clauses**

**❌ DON'T**

```python
def load_and_divide(filename, x):
    try:
        f = open(filename)
        value = int(f.read())
        return value / x
    except Exception:
        print("Something went wrong")


print(load_and_divide("data.txt", 0))
```

**✔️ DO**

```python
def load_and_divide(filename, x):
    try:
        f = open(filename)
        value = int(f.read())
        return value / x
    except FileNotFoundError:
        return "File not found"
    except ValueError:
        return "File content is not an integer"
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(load_and_divide("missing.txt", 2))
print(load_and_divide("../../think_python/data/words.txt", 0))
```
