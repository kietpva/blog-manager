# **If Statements**
## **Use if statements to return early on error conditions**

**❌ DON'T**

```python
def withdraw(balance, amount):
    if amount > 0:
        if amount <= balance:
            balance -= amount
            return balance
        else:
            return "Error: Insufficient funds"
    else:
        return "Error: Invalid amount"
print(withdraw(1000, -100))
print(withdraw(1000, 1500))
```

**✔️ Do**

```python
def withdraw(balance, amount):
    if amount <= 0:
        return "Error: Invalid amount"

    if amount > balance:
        return "Error: Insufficient funds"

    balance -= amount
    return balance
print(withdraw(1000, -100))
print(withdraw(1000, 1500))
```

## **Chain comparisons to make if statements more concise**
**❌ DON'T**


```python
def is_within_range(number, min, max):
  if min <= number and number <= max:
    return True
  return False

print(is_within_range(3, 2, 15)) 
print(is_within_range(15, 2, 10))
```

**✔️ Do**

```python
def is_within_range(number, min, max):
  if min <= number <= max:
    return True
  return False

print(is_within_range(3, 2, 15)) 
print(is_within_range(15, 2, 10))
```
