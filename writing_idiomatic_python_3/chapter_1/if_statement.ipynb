# **If Statements**
## **Use if statements to return early on error conditions**

**❌ DON'T**

```python
def withdraw(balance, amount):
    if amount > 0:
        if amount <= balance:
            balance -= amount
            return balance
        else:
            return "Error: Insufficient funds"
    else:
        return "Error: Invalid amount"
print(withdraw(1000, -100))
print(withdraw(1000, 1500))
```

**✔️ Do**

```python
def withdraw(balance, amount):
    if amount <= 0:
        return "Error: Invalid amount"

    if amount > balance:
        return "Error: Insufficient funds"

    balance -= amount
    return balance
print(withdraw(1000, -100))
print(withdraw(1000, 1500))
```

## **Chain comparisons to make if statements more concise**
**❌ DON'T**


```python
def is_within_range(number, min, max):
  if min <= number and number <= max:
    return True
  return False

print(is_within_range(3, 2, 15)) 
print(is_within_range(15, 2, 10))
```

**✔️ Do**

```python
def is_within_range(number, min, max):
  if min <= number <= max:
    return True
  return False

print(is_within_range(3, 2, 15)) 
print(is_within_range(15, 2, 10))
```

## **Avoid placing conditional branch code on the same line as the colon**
**❌ DON'T**

```python
name = 'Jeff'
address = 'New York, NY'
if name: print(name)
print(address)
```

**✔️ Do**

```python
name = 'Jeff'
address = 'New York, NY'
if name:
    print(name)
print(address)
```

## **Avoid repeating variable name in compound if statement**

**❌ DON'T**

```python
def can_drive(age, has_license):
    if age >= 18 and age <= 65 and has_license:
        return "Allowed to drive"
    else:
        return "Not allowed"

can_drive(26, True)
```

**✔️ Do**

```python
def can_drive(age, has_license):
    is_valid_age = 18 <= age <= 65

    if is_valid_age and has_license:
        return "Allowed to drive"

    return "Not allowed"

can_drive(26, True)
```

## **Avoid comparing directly to `True`, `False`, or `None`**
**❌ DON'T**

```python
def can_access(is_admin):
    if is_admin == True:
        return "Access granted"
    else:
        return "Access denied"

print(can_access(True))
print(can_access(False))
```

**✔️ Do**

```python
def can_access(is_admin):
    if is_admin:
        return "Access granted"
    return "Access denied"

print(can_access(True))
print(can_access(False))
```
