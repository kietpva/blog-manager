# **Functions**
## **Avoid using a mutable object as the `default value` for a function argument**
**❌ DON'T**

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))
```

**✔️ DO**

```python
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))
```

## **Use `return` to evaluate expressions in addition to return values**
**❌ DON'T**

```python
def is_even(n):
    if n % 2 == 0:
        result = True
    else:
        result = False
    return result

print(is_even(4))
print(is_even(7))
```

**✔️ DO**

```python
def is_even(n):
    return n % 2 == 0

print(is_even(4))
print(is_even(7))
```

## **Learn to use `keyword arguments` properly**

**❌ DON'T**

```python
def print_list(list_value, sep):
    print('{}'.format(sep).join(list_value))
the_list = ['a', 'b', 'c']
the_other_list = ['Jeff', 'hates', 'Java']
print_list(the_list, ' ')
print_list(the_other_list, ' ')
print_list(the_other_list, ', ')
```

**✔️ DO**

```python
def print_list(list_value, sep=' '):
    print('{}'.format(sep).join(list_value))
the_list = ['a', 'b', 'c']
the_other_list = ['Jeff', 'hates', 'Java']
print_list(the_list)
print_list(the_other_list)
print_list(the_other_list, ', ')
```

## **Use `*args` and `**kwargs` to accept arbitrary arguments**
**❌ DON'T**

```python
def log_event(level, message, user_id, timestamp):
    print(level, message, user_id, timestamp)

log_event("INFO", "Login", 123, "10:00")
```

**✔️ DO**

```python
def log_event(level, *args, **kwargs):
    print("Level:", level)
    print("Messages:", args)
    print("Extra info:", kwargs)

log_event(
    "INFO", "Login", "From mobile", user_id=123, timestamp="10:00", ip="127.0.0.1"
)
```

## **Learn to treat functions as values**
**❌ DON'T**

```python
def print_addition_table():
    for x in range(1, 3):
        for y in range(1, 3):
            print(str(x + y) + '\n')
def print_subtraction_table():
    for x in range(1, 3):
        for y in range(1, 3):
            print(str(x - y) + '\n')
def print_multiplication_table():
    for x in range(1, 3):
        for y in range(1, 3):
            print(str(x * y) + '\n')
def print_division_table():
    for x in range(1, 3):
        for y in range(1, 3):
            print(str(x / y) + '\n')
print_addition_table()
print_subtraction_table()
print_multiplication_table()
print_division_table()
```

**✔️ DO**

```python
import operator as op

def print_table(operator):
    for x in range(1, 3):
        for y in range(1, 3):
            print(str(operator(x, y)) + '\n')
for operator in (op.add, op.sub, op.mul, op.itruediv):
    print_table(operator)
```
