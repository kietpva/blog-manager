# **Functions**
## **Avoid using a mutable object as the `default value` for a function argument**
**❌ DON'T**

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))
```

**✔️ DO**

```python
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))
```

## **Use `return` to evaluate expressions in addition to return values**
**❌ DON'T**

```python
def is_even(n):
    if n % 2 == 0:
        result = True
    else:
        result = False
    return result

print(is_even(4))
print(is_even(7))
```

**✔️ DO**

```python
def is_even(n):
    return n % 2 == 0

print(is_even(4))
print(is_even(7))
```
