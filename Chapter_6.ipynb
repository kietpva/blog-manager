## Some functions have return values
This section explains that many Python functions return values that can be used in expressions or stored in variables. A function returns a value when it executes a `return` statement. If a function does not explicitly return a value, Python implicitly returns a special value called `None`

- void function - which perform an action like printing or drawing but do not return a value
- fruitful function - which compute and return a value

It also shows how you can use the value returned by one function as an argument to another function, and how returning a value makes your functions more flexible and reusable.

````python
import math

def circle_area(radius):
    return math.pi * radius**2

r = 5
area = circle_area(r)
print("Area:", area)
````

## And some have None
This section explains that functions in Python that don’t explicitly return a value actually return a special value called None. Even if there is no `return` statement, Python implicitly returns None when the function finishes

The section shows how you can observe this behavior by assigning the result of a function that doesn’t return anything to a variable and printing the result—it will be `None`. It also points out that None is a real value in Python, and if you mistakenly use a function that returns `None` in an expression, you can get confusing errors.

````python
def say_hello():
    print("Hello")

result = say_hello()
print(result)   # Output: None
````

## Return values and conditionals
This section explains how `return` statements work inside conditional structures (`if`, `else`). When a `return` is executed in a function, the function immediately stops and returns the given value. This means that code after the `return` in that branch does not run.

Also, if there are multiple branches, each should have its own `return` to ensure the function always returns a value. Otherwise, some paths may end without a return, and the function will implicitly return `None`.

This section demonstrates how to use conditionals within functions to decide what value to return depending on input. A key point is that a return inside an `if` or `else` stops further execution of the function.

````python
def sign(x):
    if x < 0:
        return -1
    elif x > 0:
        return 1
    else:
        return 0

print(sign(-10))   # -1
print(sign(0))     # 0
print(sign(5))     # 1
````

## Incremental development
Incremental development is a programming strategy where you write and test your code in small, manageable pieces. Instead of trying to write a full, correct function all at once, you build it step by step:

1. Start with a simple version of the function that does part of the task or returns a placeholder value
2. Test that simple version to ensure it works as expected
3. Gradually add more functionality, testing after each change
4. Continue refining until the function performs the complete task

This approach helps:
- catch errors early
- makes debugging easier because you know exactly which change introduced a problem.

The section demonstrates this with an example function (such as computing the hypotenuse). It starts with a version that always returns 0, then adds the correct logic step by step until the final version is complete and correct.

### Step 1: Simple version

````python
def hypotenuse(a, b):
    return 0

print(hypotenuse(3,4))  # Expected 5, currently 0
````

### Step 2: Compute squares

````python
def hypotenuse(a, b):
    dx = a * a
    dy = b * b
    return dx + dy

print(hypotenuse(3,4))  # 25 — still not correct, but partial progress
````

### Step 3: Complete version:

````python
import math

def hypotenuse(a, b):
    dx = a * a
    dy = b * b
    return math.sqrt(dx + dy)

print(hypotenuse(3,4))  # 5.0 — correct!
````

## Boolean functions
A boolean function is a function that returns either `True` or `False`. These functions are useful because they can be used directly in conditional statements like `if` or in expressions that expect a boolean value.

A boolean function typically uses a relational or logical expression in its `return` statement. Instead of using an `if` to explicitly return `True` or `False`, you can often simply return the result of the expression itself since comparisons already produce `True` or `False`.

This section encourages writing small, reusable boolean functions to test properties of values, such as determining whether a number is even or whether three lengths can form a triangle.

**Example**

````python
def is_even(n):
    return n % 2 == 0

print(is_even(10))  # True
print(is_even(7))   # False
````

## Recursion with return values
This section explains how recursion and return values work together. When a recursive function needs to produce a result, it must return the results of recursive calls so that each call contributes to the final answer. Unlike a recursive function that only prints values, one that returns values can be used in expressions and assigned to variables.

The key idea is that each recursive call returns a value to its caller, which then combines those values until the base case is reached. This lets you write functions that compute things like factorials or Fibonacci numbers, where each step’s result depends on smaller subproblems.

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))  # 120
````

## Fibonacci
This section explains how to write a recursive function to compute the Fibonacci numbers. The Fibonacci sequence is defined such that:
- The 0th Fibonacci number is 0
- The 1st Fibonacci number is 1
- Every subsequent number is the sum of the two preceding ones

So, The sequence starts:

```
0, 1, 1, 2, 3, 5, 8, 13, 21, ...
```
To compute the nth Fibonacci number using recursion, the function must check for the base case (`n==0` and `n == 1`), and for all other values call itself for the two smaller values and return their sum

The section also notes that this recursive implementation is inefficient because it recomputes the same value many times, but it is usful for illustrating recurion with return values.

````python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))  # 8
````
