# Chapter 6: Return Values

## Some functions have return values
This section explains that many Python functions return values that can be used in expressions or stored in variables. A function returns a value when it executes a `return` statement. If a function does not explicitly return a value, Python implicitly returns a special value called `None`

- void function - which perform an action like printing or drawing but do not return a value
- fruitful function - which compute and return a value

It also shows how you can use the value returned by one function as an argument to another function, and how returning a value makes your functions more flexible and reusable.

```python
import math

def circle_area(radius):
    return math.pi * radius**2

r = 5
area = circle_area(r)
print("Area:", area)
```

## And some have None
This section explains that functions in Python that don’t explicitly return a value actually return a special value called None. Even if there is no `return` statement, Python implicitly returns None when the function finishes

The section shows how you can observe this behavior by assigning the result of a function that doesn’t return anything to a variable and printing the result—it will be `None`. It also points out that None is a real value in Python, and if you mistakenly use a function that returns `None` in an expression, you can get confusing errors.

```python
def say_hello():
    print("Hello")

result = say_hello()
print(result)   # Output: None
```
