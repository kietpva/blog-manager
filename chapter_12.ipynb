# **Text Analysis and Generation**
## **Unique words**
This section explains how to find the number of unique words in a text file. After reading all the words from a file into a list, we want to know how many distinct words appear—i.e., words that occur at least once.

To determine unique words, we use a set in Python. A set is an unordered collection of distinct elements. By converting a list of words into a set, all duplicates are automatically removed. Then the number of unique words is simply the size (length) of the set.

In practical terms:

1. Read all words from a file into a list
2. Convert the list to a set
3. Use len() on the set to get the number of unique words

This technique use ful in analyzing text to understand vocabulary richness or to compare different texts based on their distinct word usage

### **Example**
#### **Reading words from a file**

```python
def read_words(filename):
    words = []
    fin = open(filename)
    for line in fin:
        for word in line.split():
            words.append(word)
    return words

print(read_words('words.txt'))
```

#### **Finding unique words**

```python
words = read_words('words.txt')
unique = set(words)
print(len(unique))  # number of unique words
```

## **Punctuation**
This section explains how to handle punctuation when processing text for word analysis. When reading words from a file, many words may include punctuation marks attached to them (like commas, periods, quotation marks), which can distort the analysis.

To get meaningful word counts or unique word lists, you need to remove punctuation from the text before processing. The idea is to define a set of punctuation characters and then strip them from each word so that a word like `"hello,"` becomes just `"hello"`.

A common approach:
1. Define a string of punctuation characters to remove (e.g., `string.punctuation`)
2. For each word read from the file, use methods like `strip()` or iterate through characters to filter out any punctuation
3. Word with the cleaned words for freequency counting or other analysis

This ensures that punctuation attached to words does not affect the sults (e.g., `'word'` and `'word.'` are treated as the same word)

### **Example**
#### **Python code to remove punctuation**

```python
import string

word = "hello,"
clean = word.strip(string.punctuation)
print(clean)  # "hello"
```

#### **Removing punctuation from words in a list**

```python
import string

words = ["cat,", "dog!", "it's", "hello."]
clean_words = []
for w in words:
    clean_words.append(w.strip(string.punctuation))

print(clean_words)
```
