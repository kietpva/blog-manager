# **Text Analysis and Generation**
## **Unique words**
This section explains how to find the number of unique words in a text file. After reading all the words from a file into a list, we want to know how many distinct words appear—i.e., words that occur at least once.

To determine unique words, we use a set in Python. A set is an unordered collection of distinct elements. By converting a list of words into a set, all duplicates are automatically removed. Then the number of unique words is simply the size (length) of the set.

In practical terms:

1. Read all words from a file into a list
2. Convert the list to a set
3. Use len() on the set to get the number of unique words

This technique use ful in analyzing text to understand vocabulary richness or to compare different texts based on their distinct word usage

### **Example**
#### **Reading words from a file**

````python
def read_words(filename):
    words = []
    fin = open(filename)
    for line in fin:
        for word in line.split():
            words.append(word)
    return words

print(read_words('words.txt'))
````

#### **Finding unique words**

````python
words = read_words('words.txt')
unique = set(words)
print(len(unique))  # number of unique words
````

## **Punctuation**
This section explains how to handle punctuation when processing text for word analysis. When reading words from a file, many words may include punctuation marks attached to them (like commas, periods, quotation marks), which can distort the analysis.

To get meaningful word counts or unique word lists, you need to remove punctuation from the text before processing. The idea is to define a set of punctuation characters and then strip them from each word so that a word like `"hello,"` becomes just `"hello"`.

A common approach:
1. Define a string of punctuation characters to remove (e.g., `string.punctuation`)
2. For each word read from the file, use methods like `strip()` or iterate through characters to filter out any punctuation
3. Word with the cleaned words for freequency counting or other analysis

This ensures that punctuation attached to words does not affect the sults (e.g., `'word'` and `'word.'` are treated as the same word)

### **Example**
#### **Python code to remove punctuation**

````python
import string

word = "hello,"
clean = word.strip(string.punctuation)
print(clean)  # "hello"
````

#### **Removing punctuation from words in a list**

````python
import string

words = ["cat,", "dog!", "it's", "hello."]
clean_words = []
for w in words:
    clean_words.append(w.strip(string.punctuation))

print(clean_words)
````

## **Word frequencies**
This section explains how to compute the frequency of each word in a text file — that is, how often each word appears.

The steps typically include:

1. Reading all words from the file, cleaning punctuation, and normalizing (e.g., making all letters lowercase).
2. Using a dictionary to count occurrences of each word (like value_counts from earlier chapters)
3. Converting the dictionary of word counts into a list of (count, word) pairs
4. Sorting that list so that you can identify the most frequent words

This process is a basic text analysis technique for understanding how often words occur in a file, such as a novel or an article.

### **Example**
#### **Read and clean words**

````python
import string

def read_words(filename):
    words = []
    with open(filename) as fin:
        for line in fin:
            for raw in line.split():
                word = raw.strip(string.punctuation).lower()
                if word != "":
                    words.append(word)
    return words

print(read_words('words.txt'))
````

#### **Count word frequency**

````python
def word_frequencies(words):
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    return counts

print(word_frequencies(read_words('words.txt')))
````

#### **Convert dictionary to list of**

````python
def invert_counts(counts):
    items = []
    for word, count in counts.items():
        items.append((count, word))
    return items

print(invert_counts(word_frequencies(read_words('words.txt'))))
````

#### **Sort and display top frequent words**

````python
items = invert_counts(word_frequencies(read_words('words.txt')))
items.sort(reverse=True)
for count, word in items[:10]:
    print(word, count)
````

## **Optional parameters**
This section explains how functions in Python can have optional parameters — parameters that have default values if no argument is provided.

When you define a function, you can assign a default value to a parameter by using the equals sign (`=`):

```python
def func(arg1, arg2=default_value):
```
if the caller doesn't provide a value for that parameter, Python uses the default. This is useful for defining function where some arguments are commonly omitted or where a reasonable default can be assumed

For example, in word frequency analysis, you might want to create a function that prints the most frequent words, but if the user doesn't specify how many to print, a default number (like 10) is used.

Optional parameters should be listed after required ones. You can combine multiple optional parameters, and default values can be of any type.

### **Example**
#### **Function with an optional parameter**

````python
def repeat(word, times=2):
    result = word * times
    return result

print(repeat('hello '))
````

## **Dictionary subtraction**
This section introduces the concept of dictionary subtraction, which means removing keys from one dictionary based on the contents of another dictionary. In the context of word frequency analysis, you might want to remove common words (stop words) from a dictionary of word counts to focus on more meaningful terms.

The idea is:
- Given two dictionaries (`big` and `small`), remove from `big` all keys that also appear in `small`.
- This process leaves only the keys in `big` that are not present in `small`.

A common implementation pattern is to loop over the keys of the first dictionary and add to a new dictionary only those keys that are not in the second dictionary.

This “dictionary subtraction” is useful when filtering data — for example, removing “stop words” from a frequency count.

### **Example**

````python
def subtract(d1, d2):
    result = {}
    for key in d1:
        if key not in d2:
            result[key] = d1[key]
    return result

counts = {'the': 50, 'and': 30, 'python': 10, 'code': 5}
stop_words = {'the':1, 'and':1}
filtered = subtract(counts, stop_words)
print(filtered)
````

## **Bigrams**
This section introduces the concept of bigrams, which are pairs of consecutive items in a sequence — usually words or characters. Bigrams are a basic tool in text analysis and language processing because they capture relationships between elements (e.g., how often one word follows another).

**Word bigrams**

A word bigram is formed by taking pairs of consecutive words in a text. For example, in the sentence:
```python
I love Python programming
```
the word bigrams are:
```
('I', 'love')
('love', 'Python')
('Python', 'programming')
```

**Character bigrams**
Similarly, a character bigram comes from consecutive characters in a string. For example, in "word", the character bigrams are:
```
('w','o'), ('o','r'), ('r','d')
```
To extract bigrams from a list of words, you can loop over adjacent pairs. Bigrams are useful when building frequency models (such as how often a given bigram appears) and are foundational in many probabilistic language models.
### **Example**
#### **Word bigrams from a list**

````python
def word_bigrams(words):
    pairs = []
    for i in range(len(words) - 1):
        pairs.append((words[i], words[i+1]))
    return pairs

words = ['I', 'love', 'Python', 'programming']
print(word_bigrams(words))
````

#### **Character bigrams from a word**

````python
def char_bigrams(word):
    return [(word[i], word[i+1]) for i in range(len(word) - 1)]

print(char_bigrams('word'))
````

## Markov analysis
This section introduces the idea of Markov analysis applied to text. Markov analysis looks at sequences of items (such as words) and tries to understand the probability of the next item based on the current item or a state.

In text processing, a simple Markov model can be built by looking at bigrams (pairs of consecutive words). For example, if the sequence is `w1, w2, w3`, ..., then for each `w_i` you record which word follows `w_i` in the text. From this you can build a mapping (dictionary) where:

- **key** = a word
- **value** = a list of words that have followed the key in the text

This dictionary of bigram links can then be used to generate random text by choosing a starting word and randomly picking one of the possible next words, then repeating.

The idea is foundational in simple text generation and shows how you can use observed probabilities in a corpus to simulate plausible sequences.

### **Example**
#### **Build the mapping**

````python
def build_markov(words):
    markov = {}
    for i in range(len(words) - 1):
        key = words[i]
        next_word = words[i+1]
        if key not in markov:
            markov[key] = []
        markov[key].append(next_word)
    return markov

words = ['I', 'love', 'Python', 'programming']
print(build_markov(words))
````

#### **Use the model**

````python
import random

def generate_text(markov, start, n):
    current = start
    result = [current]
    for _ in range(n):
        next_words = markov.get(current)
        if not next_words:
            break
        current = random.choice(next_words)
        result.append(current)
    return ' '.join(result)

print(generate_text(build_markov(words), 'I', 10))
````
