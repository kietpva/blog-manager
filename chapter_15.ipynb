# **Classes and Methods**
## **Defining methods**
To turn a regular function (like `print_time`) into a method, move it inside the class indent and give it a first parameter named `self`

```python
class Time:
    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)

# call method
t = Time()
t.hour = 9; t.minute = 40; t.second = 1

t.print_time()
```

## **Another method**
Can you convert the `time_to_int` function into a method

```python
class Time:
    def time_to_int(self):
        return self.hour * 3600 + self.minute * 60 + self.second

# call
t = Time()
t.hour = 1
t.minute = 2
t.second = 3

print(t.time_to_int())  # 3723
```

## Static methods
Static methods don't use `self` and can be called on the class it self, useful for operations that create new instances

```python
class CustomTime:
    def __init__(self, hour, minute, second):
        self.hour = hour
        self.minute = minute
        self.second = second

    @staticmethod
    def from_string(time_str):
        hour, minute, second = map(int, time_str.split(':'))
        return CustomTime(hour, minute, second)

t = CustomTime.from_string("09:45:30")

print(t.hour)
print(t.minute)
print(t.second)
```

## **Comparing Time objects**
Can you write an is_after method that returns True/False when one time is after another? For example:

```python
class Time:
    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def time_to_int(self):
        return self.hour * 3600 + self.minute * 60 + self.second

    def is_after(self, other):
        return self.time_to_int() > other.time_to_int()

start = Time(9, 30, 0)
end = Time(10, 15, 0)

print(end.is_after(start))   # True
print(start.is_after(end))   # False
```
