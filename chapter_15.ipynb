# **Classes and Methods**
## **Defining methods**
To turn a regular function (like `print_time`) into a method, move it inside the class indent and give it a first parameter named `self`

```python
class Time:
    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)

# call method
t = Time()
t.hour = 9; t.minute = 40; t.second = 1

t.print_time()
```

## **Another method**
Can you convert the `time_to_int` function into a method

```python
class Time:
    def time_to_int(self):
        return self.hour * 3600 + self.minute * 60 + self.second

# call
t = Time()
t.hour = 1
t.minute = 2
t.second = 3

print(t.time_to_int())  # 3723
```
