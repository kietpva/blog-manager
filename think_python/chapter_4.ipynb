## Making a square
The _Making a square_ section shows to draw a square using the jupyturtle module. It starts by using repeated calls to `forward()` and `left()` to move a turtle the same distance in a square pattern. Because the four pairs of statements are identical, the example then shows how to use a `for` loop to make the code more concise and a void repetition. This illustrates how loops can simplify repetitive patterns in code.

**Key Idea**
- A square has four equal sides and four right-angle turns
- Instead of writing the same line four times, we use a for loop to repeat it
- This highlights how loops make code cleaner and reduce repetition

````python
make_turtle()

for i in range(4):
    forward(50)
    left(90)
````

This code creates a turtle and moves it forward 50 units, then turns left 90 degrees, repeated four times - drawing a square

## Encapsulation and Generalization

1. Encapsulation means wrapping a sequence of statements into a function and giving it a name. This makes the code easier to reuse and gives it documentation-like clarity. For example, instead of repeating the same forward/turn statements to draw a square, you can put them into a function named
2. Initially, the fuction might draw a square of fixed size. To make it more useful, generalization adds parameters to the function so that it can draw squares of different sizes. That means replacing fixed values(like side length 50) with parameters(like `length`)
3. Youcan generalize further: a function does not have to draw only squares. For example, replaceing the hard-coded value "4 sides" with a parameter `n` lets u draw any regular polygon using a function like `polygon(n, length)

This process - first encapsulating repetitive code into a function, then generalizing it with parameters - is a key

**Example:**

````python
# Encapsulation: wrap code into a function
def square():
    for i in range(4):
        forward(50)
        left(90)

make_turtle()
square()

# Generalization: add a parameter
def square(length):
    for i in range(4):
        forward(length)
        left(90)

square(30)
square(60)

# Further generalization: polygon with any number of sides
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)

polygon(7, 30)   # 7-sided shape
````

## Approximating a circle
In Approximating a circle, the book explains how to draw a circle approximately using the polygon function you already wrote. A circle can be approximated by drawing a polygon with many small sides, so the edges blend together visually.

To do this, you:
1. Compute the circumference of the circle using the formula:
    ```python
    circumference = 2 * math.pi * radius
    ```
2. Choose a number of segments 'n' to divide the circle into. A larger `n` makes the shape closer to a true circle.
3. Compute the length of each segment:
   ```python
   length = circumference / n
   ```
4. Call the `polygon` function with these values so that it draws an `n`-sided shape approximating the circle.

````python
# import math

# def circle(t, r):
#     circumference = 2 * math.pi * r
#     n = 50
#     length = circumference / n
#     polygon(t, n, length)

from jupyturtle import make_turtle, forward, left
import math

def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)

def circle(r):
    circumference = 2 * math.pi * r
    n = 50
    length = circumference / n
    polygon(n, length)

make_turtle(delay=0.02)
circle(30)
````

The idea is that you reuse the polygon-drawing code to draw something that looks like a circle. A limitation is that if `n` is fixed (like 50), large circles have long straight segments and small circles waste time drawing too many tiny segments. To improve this, you could make `n` dependent on the circle's size, but the simple version keeps the interface clean

## Refactoring
The Refactoring section explains how to reorganize code to make it clearer and more general without changing its behavior. The example starts with separate functions that have duplicate logic, such as a `polygon` function and an `arc` function that both draw many short line segments.

Instead of repeating the same loop logic in different functions, you can extract the common pattern into another helper function. In this case, a function named polyline is introduced that takes the number of segments, the length of each segment, and the turning angle as parameters. The `polygon` and `arc` functions can then call `polyline`, which reduces duplication and makes the code easier to read and maintain.

This process — identifying common code, extracting it into a shared function, and rewriting original functions in terms of the helper — is called refactoring. It improves code structure without altering what the program does. Refactoring is a key practice in writing clean, maintainable software.

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)

def polygon(n, length):
    angle = 360 / n
    polyline(n, length, angle)

def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)

def circle(radius):
    import math
    circumference = 2 * math.pi * radius
    n = 50
    length = circumference / n
    polyline(n, length, 360/n)

make_turtle(delay=0.005)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

## A development plan
The A development plan section lays oit a systematic approach to building programs by breaking the task into smaller, manageable steps. Rather than writing a complete program all at once, the idea is to start with a simple version and gradually improve it by introducing functions, generalizing parameters and refactoring repeated code.

The steps in the development plan are:

1. Write a short, simple program that computes the result you want using only basic statements
2. Encapsulated: Identify repeated code and group it into functions
3. Generalize: Add parameters so functions can handle a wider range of inputs
4. Refactor: once the basic version works, clean up the code by removing duplication and improving structure

This approach helps you make steady progress, test freqyebtly, and avoid getting stuck trying to write the perfect solution on the first try. It mirrors how programmers gradually refine their programs in practice

## Docstrings
A docstring is a string at the beginning of a function definition that explains the function’s interface — including what the function does and what its parameters mean. It appears inside triple quotes (`"""`) so it can extend across multiple lines.

A docstring should:
- Describe the purpose of the function without explaining the implementation details.
- Describe the meaning and effect of each parameter.
- Indicate the expected type of each parameter if not obvious.

Writing good docstrings is an important part of interface design. If a function’s interface is hard to explain, it may indicate the interface needs to be improved.

**Example**

````python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """
    for i in range(n):
        forward(length)
        left(angle)
````

Write a function called `rectangle` that draws a rectangle with given side lengths. For example, here’s a rectangle that’s `80` units wide and `40` units tall.

````python
from jupyturtle import make_turtle, forward, left

def rectangle(t, width, height):
    """Draws a rectangle with the given width and height."""
    for i in range(2):
        forward(width)
        left(90)
        forward(height)
        left(90)

t = make_turtle()
rectangle(t, 80, 40)
````

Write a function called `rhombus` that draws a rhombus with a given side length and a given interior angle. For example, here’s a rhombus with side length `50` and an interior angle of `60` degrees.

````python
from jupyturtle import make_turtle, forward, left

def rhombus(t, length):
    """Draws a rhombus with the given width and height."""
    for i in range(2):
        forward(length)
        left(60)
        forward(length)
        left(120)

t = make_turtle()
rhombus(t, 50)
````
