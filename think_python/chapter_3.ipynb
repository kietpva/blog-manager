# Function
A function is a named sequence of statements that performs a computation. When you define a function, the statements inside it do not run until the function is called.

## Defining new functions
You define a function using the *def* keyword, followed by the function name, parentheses, and a colon. The body of the function must be indented.

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")

print_lyrics()
````

## Parameters and arguments
Functions can take parameters, which are variables inside the function. When calling a function, you pass arguments to these parameters.

````python
def greet(name):
    print("Hello", name)

greet("Python")
````

- `name` is a parameter
- `Python` is an argument

## Calling functions
Once a function is defined, you can call it by using its name followed by parentheses. A function can also call another function.

````python
def line():
    print("----------")

def message():
    line()
    print("Welcome!")
    line()

message()
````

- Function `message` call function `line`

## Repetition
Repetition means doing the same set of actions multiple times. In Python, you use a `for` loop to repeat code when you know how many times it should run. A `for` loop has a header that starts with `for`, a loop variable, the keyword `in`, and something that produces a sequence of values. The body of the loop (the indented block) runs once for each value in the sequence. For example:

````python
for i in range(2):
    print(i)
````

The `range(2)` produces the values `0` and `1`. Each time through the loop, `i` gets the next value and the body runs again. You can use a `for` loop to repeat any block of statements, including calling other functions like printing verses of a song multiple times.

The loop is called a loop because after executing the body once, Python goes back (“loops back”) to get the next value and runs the body again until there are no more values.

## Variables and Parameters are Local
When you create a variable inside a function, that variable is local to the function. This means:
- The variable only exists inside the function where it is defined.
- After the function finishes running (returns), the local variable is destroyed and you cannot use it outside the function.

````python
def print_twice(s):
    print(s)

def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
print(cat)
````

Here, `cat` is a local variable: it exists only inside `cat_twice`. Attempting to access cat outside will cause an error (`NameError`).

Additionally, function parameters (like `part1` and `part2`) are also local. They behave like local variables and cannot be accessed outside the function body.

## Stack Diagrams
A stack diagram is a way to visualize what happens when functions call each other and how variables are stored during execution. Like state diagrams, stack diagrams show the values of variables, but they also show which function each variable belongs to.
- Each function call creates a frame.
  - A *frame* is drawn as a box labeled with the function’s name.
  - Inside the frame are the parameters and local variables for that call.
- Frames are arranged in a stack.
  - The bottom of the stack is the function currently running, and above it are the functions that called it, and so on.
  - For example, if `print_twice` was called by `cat_twice`, and `cat_twice` was called by the main program (`__main__`), the stack reflects that order.
- When a variable is created outside any function, it belongs to `__main__`.
- Each parameter refers to the same value as the corresponding argument passed when the function was called.

| Stack order | Shows which function called which function, arranged as a stack. |
| Local variables & parameters | Stored inside the frame of the function that created them. |

## Tracebacks
A traceback is the error message Python prints when a program crashes due to an exception.
It shows the sequence of function calls that led to the error, starting from `__main__` and ending at the line where the error occurred.

The last line of the traceback tells you what the error is.

### Example 1

````python
def print_twice(s):
    print(s)
    print(s)

def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)

def main():
    print_twice(cat)

main()
````

**Explanation**
- `cat` is a local variable of `cat_twice`, not visible in `main`.

### Example 2

````python
def divide(a, b):
    return a / b

def compute():
    result = divide(10, 0)
    print(result)

compute()
````

**Explanation**
- The traceback shows every function call leading to the error.

### How to Read a Traceback
- Read from the bottom up
- Identify the error type
- Look at the line number
- Check the function scope

**Note:** _The bottom line of a traceback tells you what when wrong._
_The lines above tel you how Python got there._

## Why functions?
It might not be obvious why it’s worth dividing a program into functions, but there are several good reasons:
- **Improve readability:** Creating a new function gives you chance to give a name to a group of statements, which makes your program easier to read and debug
- **Reduce repetition:** Functions help eliminate repetitive code. If you need to change something, you only have to change it in one place.
- **Simplify debugging:** Dividing a long program into functions allows you to debug parts separately and then assemble them into a working whole.
- **Reusability:** Well-designed functions can often be used in many different programs. Once you write and debug a function, you can reuse it elsewhere.

## Debugging
Debugging is the process of finding and fixing errors in a program. It can sometimes be frustrating, but it is also interesting and rewarding — and one of the most important skills you can learn.

**Key ideas:**
- Debugging is like detective work: you gather clues from the program’s behavior and try to infer what went wrong.
- It is also like experimental science: form a hypothesis about what is causing the error, then change the code and test it. If your idea was correct, you’ll see predictable improvements; if not, you try again with a new hypothesis.
- Some programmers think of programming itself as debugging: you start with a program and gradually refine it until it does what you want.
- If you spend too much time debugging, it often means you wrote too much code at once without testing. Taking smaller steps and testing often usually leads to faster progress.

1. Write a function named `print_right` that takes a string named `text` as a parameter and prints the string with enough leading spaces that the last letter of the string is in the 40th column of the display.

    Hint: Use the `len` function, the string concatenation operator (`+`) and the string repetition operator (`*`).

````python
def print_right(text):
    space = 40 - len(text)
    print(' ' * space + text)

print_right("Hello")
print_right("Python")
````

2. Write a function called `triangle` that takes a string and an integer and draws a pyramid with the given height, made up using copies of the string. Here’s an example of a pyramid with `5` levels, using the string `'L'`.

````python
def triangle(text, height):
    for i in range(1, height + 1):
        chars = 2 * i - 1
        print(text * chars)

triangle('L', 5)
````

3. Write a function called `rectangle` that takes a string and two integers and draws a rectangle with the given width and height, made up using copies of the string. Here’s an example of a rectangle with width `5` and height `4`, made up of the string `'H'`.

````python
def rectangle (text, width, height):
    for i in range(1, height + 1):
        print(text * width)

rectangle('H', 5, 4)
````
