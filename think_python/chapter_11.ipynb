# **Tuples**
## **Tuples are like lists**
In Python, a tuple is a built-in sequence type that is very similar to a list. Like lists:

- Tuples contain a sequence of value
- The values can be of any type
- You can use indexing and slicing to access elements
- You can use sequence operators such as `+` (concatenation) and `*` (repeat)
- Functions like `len()` and `reversed()` work with tuples

However, unlike lists, tules are immutable, which means that once a tuple is created, yoi can't change its elements(more on that in the next section). Because tuples are immutable, they can be used as keys in dictionaries and in other contexts where immutablility is required

### **Example**
#### **Creating a tuple**

````python
t = ('a', 'b', 'c')
print(t)          # ('a', 'b', 'c')
print(type(t))    # <class 'tuple'>
````

#### **Indexing and slicing like lists**

````python
t = ('p','y','t','h','o','n')
print(t[0])       # 'p'
print(t[1:4])     # ('y', 't', 'h')
````

#### **Concatenate and repeat**

````python
t1 = ('a','b')
t2 = ('c','d')
print(t1 + t2)    # ('a','b','c','d')
print(t1 * 3)     # ('a','b','a','b','a','b')
````

#### **Tuple from sequence**

````python
print(tuple('hello'))   # ('h','e','l','l','o')
````

## **But tuples are immutable**
This section explains that although tuples are similar to lists in many ways, there is a fundamental difference: tuples are immutable. This means once a tuple is created, you cannot change its contents. You cannot assign to an element, insert or delete elements, or modify a slice.

```python
t = (1, 2, 3)
t[0] = 100      # this causes an error
```

The above will raise a `TypeError` because tuples do not support item assignment

The reason tuples are immutable is that it allows Python to optimize their storage and use them as keys in dictionaries. Since immutability guarantees that the data won't change, Python can safely use the has of a tuple for fast lookups

This immutability is similar to strings: you can slice a tuple and get a new tuple, but you cannot modify the original tuple

### **Example**
#### **Trying to mutate a tuple**

````python
t = (1, 2, 3)
t[0] = 10    # TypeError!
````

#### **Slice creates a new tuple**

````python
t = (1, 2, 3, 4)
print(t[1:3])   # (2, 3)
````

## **Tuple assignment**
This section introduces a convenient feature in Python called tuple assignment. Instead of assigning each variable separately, Python allows you to assign multiple variables at once using a tuple on the left side of the assignment.

```python
x, y = 1, 2
```
Here, `x` becomes `1` and `y` becomes `2`. You can also assign the result of a function that returns a tuple:
```python
t = (3, 4)
x, y = t
```
This assigns `x = 3`, `y = 4`

Tuple assignment is especially useful when a function returns a tuple. You can unpack the tuple directly into multiple variables without indexing

It also works for swapping values:
```python
x, y = y, x
```
Python evaluates the right-hand side first, then assigns to the left-hand variables
### **Example**
#### **Multiple assignment**

````python
a, b, c = 1, 2, 3
print(a)   # 1
print(b)   # 2
print(c)   # 3
````

#### **Unpacking a tuple**

````python
t = (10, 20)
x, y = t
print(x, y)   # 10 20
````

#### **Swapping variables**

````python
x = 5
y = 7
x, y = y, x
print(x, y)   # 7 5
````

## **Tuples as return values**
This section explains that functions in Python can return multiple values by returning a tuple. Instead of returning a single value, you can return a tuple containing several values.

Python makes this easy because:
- You can write `return x, y, z` without parentheses — Python automatically packs the values into a tuple.
- When the function is called, you can unpack the returned tuple into variables using tuple assignment.

This is useful when function needs to provide more than one piece of infomation. For example, you might want to return both the quotient and remainder of a division, or both the minimum and maximum of a list.

### **Example:**
#### **Returning a tuple from a function**

````python
def min_max(values):
    return min(values), max(values)
````

#### **Unpack return values**

````python
small, large = min_max([3, 7, 1, 5])
print(small)   # 1
print(large)   # 7
````

#### **Example with quotient & remainder**

````python
def divide(x, y):
    return x // y, x % y

q, r = divide(17, 5)
print(q, r)  # 3 2
````

## **Argument**
### **Argument Packing**
When you define a function with a parameter prefixed by `*`, like `*args`, Python collects all extra positional arguments into a tuple. This is called packing because it packs multiple arguments into a single tuple

```python
def f(*args):
    print(args)
```

calling `f(1, 2, 3)` packs the three arguments into the tuple (1, 2, 3)

### **Argument unpacking**
Conversely, if you have a tuple of values and want to pass each element as a separate argument to a function, you can use the `*` operator to unpack the tuple:
```python
t = (1, 2, 3)
f(*t)
```
python expands the tuple into separate arguments: `f(1, 2, 3)`

Argument packing and unpacking make functions flexible for handling variable numbers of arguments

### **Example**
#### **Argument packing**

````python
def show_all(*args):
    print(args)

show_all(10, 20, 30, 40)
````

#### **Argument unpacking**

````python
def add(x, y, z):
    return x + y + z

numbers = (1, 2, 3)
print(add(*numbers))
````

## **Zip**
This section introduces the built-in function zip(), which combines two (or more) sequences element-by-element into a sequence of tuples. Each tuple contains one element from each input sequence at the same position.

The basic idea:
- `zip(a, b)` takes two sequences `a` and `b` of the same (or different) length
- It returns an iterator of tuples `(a[i], b[i])` for each index i until the shortest sequence ends.
- You can convert the result to a list using `list(zip(...))`.

`zip()` is useful when you want to loop through two sequences in parallel without using indicaes

Example use case:
- Pairing corresponding elements
- iterating two lists together
- creating dictionary from keys nd values

### **Example**
#### **Basic use of zip**

````python
a = [1, 2, 3]
b = ['x', 'y', 'z']

pairs = list(zip(a, b))
print(pairs)  # [(1, 'x'), (2, 'y'), (3, 'z')]
````

#### **Looping with zip**

````python
for num, letter in zip(a, b):
    print(num, letter)
````

#### **Different lengths**

````python
a = [1, 2]
b = ['x', 'y', 'z']

print(list(zip(a, b)))
````

#### **Creating a dictionary**

````python
keys = ['name', 'age']
values = ['Alice', 30]
d = dict(zip(keys, values))
print(d)  # {'name': 'Alice', 'age': 30}
````

## **Comparing and sorting**
This section explains how Python compares tuples and how that behavior can be used in sorting. When Python compares tuples, it performs lexicographical ordering, meaning it compares element by element from left to right:
1. Compare the first elements
2. If they differ, that determines the result.
3. If they are equal, compare the second elements.
4. Continue until a difference is found or the tuples end.

Because of this, tuples can be sorted directly using built-in function like `sorted()` or the list method `sort()`. Python first sorts by the first element of each tuple, then by the second if the firsts are equal, and so on.

This behavior is especially useful when tuples are used as keys in lists (for example, sorting `(count, word)` pairs to order words by frequency)

### **Example**
#### **Comparing tules**

````python
print((1, 2, 3) < (2, 0, 0))  # True
````

#### **If the first elements are equal**

````python
print((1, 5, 0) < (1, 4, 9))  # False
````

#### **Sorting list of tuples**

````python
pairs = [(2, 'apple'), (1, 'banana'), (3, 'cherry')]
pairs.sort()
print(pairs)
````

## **Inverting a dictionary**
This section explains how to invert a dictionary—that is, create a new dictionary where the keys become the values and the values become the keys.

The basic idea:

- If the original dictionary maps `key → value`, the inverted dictionary should map `value → key`.
- However, this only works if the values in the original dictionary are unique and usable as dictionary keys (i.e., hashable).

A common pattern is to loop over the items of the original dictionary and assign them to the inverted dictionary:

```python
inverse = {}
for key, value in d.items():
    inverse[value] = key
```
### **Example**
#### **Basic inversion**

````python
d = {'a': 1, 'b': 2, 'c': 3}

inverse = {}
for key, value in d.items():
    inverse[value] = key

print(inverse)
````

#### **When values aren't unique**

````python
d = {'a': 1, 'b': 2, 'x': 1}

inverse = {}
for key, value in d.items():
    inverse[value] = key

print(inverse)
````

#### **Handling diplicate values**

````python
d = {'a': 1, 'b': 2, 'x': 1}
inverse = {}

for key, value in d.items():
    inverse.setdefault(value, []).append(key)

print(inverse)
````
