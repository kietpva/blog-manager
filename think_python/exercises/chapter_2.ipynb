## **What about `17 = n`?**
This statement is not legal. The left side of an assignment must be a variable, not a constant
```python
17 = n
```
This causes a `SyntaxError`

## **How about `x = y = 1`?**
This statement is legal. It assigns the value `1` to both `x` and `y`

## **What happens if you put a semi-colon (`;`) at the end of a Python statement?**
The statement is legal, but the semi-colon is unnecessary in Python
```python
n = 17;
```
Python allows `;`, but they are rarely used

## **What if you put a period (`.`) at the end of a statement?**
This causes a syntax error
```python
n = 'python'.
```
*But when a statement ends with an integer, it is interpreted as a float*
```python
n = 17.

print(type(n)) # <class 'float'>
```
