## Loops and strings
This section explains how `for` loops work with strings. In Python, a string is a sequence of characters, and a `for` loop can iterate over each character in a string. Each iteration assigns the next character to the loop variable.

A common pattern is to use a `for` loop with a string to analyze or build a new string. Because strings are immutable, you often accumulate a result in a new string or a list.

Examples illustrate looping through all characters in a string, counting occurrences of a specific character, and creating a new string with modified content.

**Example**

```python
word = "banana"
count = 0

for letter in word:
    if letter == "a":
        count = count + 1

print(count)  # 3
```

This counts how many times `a` appears in `banana`

**Example loop to build a new string**

```python
word = "think"
result = ""

for letter in word:
    result = result + letter + "."
print(result)  # t.h.i.n.k.
```

## Reading the word list

This section explains how to read a list of words from a file and store them in a Python list for further processing, such as checking whether a word is in the list or analyzing patterns.

The example uses a file called `words.txt`, which contains one word per line. To read the file:

1. Open the file using `open()`.
2. Loop through each line in the file with a `for` loop.
3. Strip off the newline (`\n`) at the end of each line using `strip()` or `rstrip()` before adding the word to a list.

After reading, you get a list called `wordlist` that contains all the words. This list can then be passed to other functions for analysis.

The section also introduces a helper function (often called `read_words()` or similar) to encapsulate this logic and return the list of words.

**Example:**

```python
def read_words(filename):
    """Return a list of words from a file."""
    wordlist = []
    fin = open(filename)
    for line in fin:
        word = line.strip()  # remove newline
        wordlist.append(word)
    return wordlist

words = read_words('words.txt')
print(words[:10])  # show first 10 words
```

## Updating variables

In Python, you often need to change (update) the value stored in a variable. This section explains how to update variables using assignment and simple arithmetic operations.

The most common pattern is to increment a variable inside a loop. For example:

```python
count = 0
for i in range(10):
    count = count + 1
```

Each time through the loop, `count` gets increased by 1. This is called an accumulator pattern, where you accumulate a result as you iterate.

Python provides a shorthand for updating variables using augmented assignment operators such as `+=`, `-=`, `*=`, etc. So the increment above can be written as:

```python
count += 1
```

## Looping and Counting
This section introduces a common programming pattern: looping through a sequence while keeping a count of something. The idea is to use a loop to visit each item in a list, string, or other sequence, and use a variable to count how many times a certain condition is met.

The typical structure includes:
1. Initialize a counter variable before the loop
2. Loop through the sequence
3. Update the counter inside the loop when a condition is true
4. Use the final counter value after the loop ends

This pattern is often used to count occurrences of a specific item or property in a sequence

**Example**

```python
words = ["apple", "banana", "cherry", "avocado"]
count = 0

for w in words:
    if len(w) > 5:
        count += 1

print(count)  # 2
```

## The `in` operator
The `in` operator is used to check whether a value exists in a sequence such as a string, list, tuple, etc. It returns a boolean (`True` or `False`). When used in a conditional like an `if` statement, it allows you to easily test membership without writing an explicit loop.

Common usage:
- Check if a character is in a string
- Check if an item is in a list
- Can be nested inside conditionals or loops

Using `in` makes programs shorter and clearer than using a manual loop to check membership

**Example**

```python
word = "banana"

if "a" in word:
    print("Yes, 'a' is in the word")
else:
    print("No, 'a' is not in the word")
```

```python
numbers = [1, 3, 5, 7, 9]

if 5 in numbers:
    print("5 is in the list")
else:
    print("5 is not in the list")
```

## Search
This section explains how to use loops and conditionals to search through a sequence (such as a string or a list) for a particular item or pattern. A typical search involves:
1. Iterating through the sequence with a for loop.
2. Using an `if` statement inside the loop to check whether the current element meets a condition.
3. If a match is found, the function can return a result immediately or take some action.
4. If the loop finishes without finding anything, it can return a default value (such as `False`, `None`, or a message).

The section highlights patterns for searching, such as checking whether a particular character is in a word, and introduces the idea of returning early from a function when a match is found.

**Example: Search for a character in a string**

```python
def has_a(word):
    for letter in word:
        if letter == 'a':
            return True
    return False

print(has_a('banana'))  # True
print(has_a('cherry'))  # False
```

**Example: Searching in a List**

```python
def find_word(words, target):
    for w in words:
        if w == target:
            return True
    return False

words = ['apple', 'banana', 'cherry']
print(find_word(words, 'banana'))  # True
print(find_word(words, 'durian'))  # False
```

## Doctest
Doctest is a testing framework in Python that allows you to write tests inside a function’s docstring. The tests look like interactive Python sessions (`>>>`) and show the expected output.

When you run `doctest`, Python executes the examples in the docstring and checks whether the actual output matches the expected output. This approach helps ensure that:
- Your code works as intended
- Your documentation statys accurate

Doctest is espcially use full for small functions and simple example

**Example with Doctest**

```python
import doctest

def add(a, b):
    """
    Returns the sum of a and b.

    >>> add(2, 3)
    5
    >>> add(-1, 1)
    0
    """
    return a + b

doctest.testmod()
```

Here’s an outline of the function that includes two doctests. Fill in the function so it passes these tests, and add at least one more doctest.

```python
import doctest

def uses_none(word, forbidden):
    """
    Checks whether a word avoids forbidden letters.

    >>> uses_none('banana', 'xyz')
    True
    >>> uses_none('apple', 'efg')
    False
    >>> uses_none('hello', 'aeiou')
    False
    >>> uses_none('sky', 'aeiou')
    True
    """
    for letter in word:
        if letter in forbidden:
            return False
    return True

doctest.testmod()
```

### Exercise 2:
Write a function called `uses_only` that takes a word and a string of letters, and that returns `True` if the word contains only letters in the string.

Here’s an outline of the function that includes two doctests. Fill in the function so it passes these tests, and add at least one more doctest.

```python
import doctest

def uses_only(word, available):
    """
    Checks whether a word uses only the available letters.

    >>> uses_only('banana', 'ban')
    True
    >>> uses_only('apple', 'apl')
    False
    >>> uses_only('', 'abc')
    True
    """
    for letter in word:
        if letter not in available:
            return False
    return True

doctest.testmod()
```
