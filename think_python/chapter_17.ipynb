## **Representing Cards**
The chapter begins by showing how to represent playing cards in Python using a class with `suit` and `rank` attributes.

````python
class Card:
    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return f"{Card.rank_names[self.rank]} of {Card.suit_names[self.suit]}"
    
    def __eq__(self, other):
        return self.suit == other.suit and self.rank == other.rank
    
    def __lt__(self, other):
        return (self.suit, self.rank) < (other.suit, other.rank)

queen = Card(1, 12)
print(queen)   # Queen of Diamonds
````

````python
cards = [
    Card(3, 13),  # King of Spades
    Card(0, 1),   # Ace of Clubs
    Card(1, 12),  # Queen of Diamonds
]

cards.sort()

for c in cards:
    print(c)
````

**Trick**

- `lt` -> less than -> `<`
- `gt` -> greater than -> `>`
- `le` -> less equal -> `<=`
- `ge` -> greater equal -> `>=`
- `eq` -> equal -> `==`
- `ne` → not equal -> `!=`

## **Decks**
Defines a `Decks` class that contains a list of cards, and a static method to generate all 52 cards

````python
import random

class Deck:
    def __init__(self, cards):
        self.cards = cards

    @staticmethod
    def make_cards():
        cards = []
        for suit in range(4):
            for rank in range(1, 14):
                cards.append(Card(suit, rank))
        return cards

    def __str__(self):
        lines = [str(card) for card in self.cards]
        return "\n".join(lines)

    def take_card(self):
        return self.cards.pop()
    
    def put_card(self, card):
        self.cards.append(card)
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def sort(self):
        self.cards.sort()
    
    def move_cards(self, other, num):
        for i in range(num):
            card = self.take_card()
            other.put_card(card)


cards = Deck.make_cards()
deck = Deck(cards)
print(len(deck.cards))  # 52
````

## **Add, remove, shuffle and sort**
This section shows how a `Deck` object can behave like a real deck of cards:
- take a card
- put a card back
- shuffle cards randomly
- sort cards using comparison rules

### **Take a card(remove)**
`take_card()` removes and returns on card from the deck. by default, it removes the last card using `pop()`

```python
class Deck:
    ...
    def take_card(self):
        return self.cards.pop()
```

````python
deck = Deck(Deck.make_cards())

card = deck.take_card()
print(card)
print(len(deck.cards))
````

### **Put a card (add)**
`put_card()` adds a card to the deck using `append()`

```python
class Deck:
    ...
    def put_card(self, card):
        self.cards.append(card)
```

````python
deck = Deck([])
card = Card(3, 1)   # 2 of Spades

deck.put_card(card)
print(len(deck.cards))  # 1
print(deck.cards[0])
````

### **Shuffle the deck**
`shuffle()` randomizes the oder o card using `random.shuffle`

```python
import random

class Deck:
    ...
    def shuffle(self):
        random.shuffle(self.cards)
```

````python
deck = Deck(Deck.make_cards())

print(deck.cards[0])
deck.shuffle()
print(deck.cards[0])
````

### **Sort the deck**
`sort()` orders the cards using Python's list `sort()` method. This works because `Card` has `__It__`

```python
class Deck:
    ...
    def sort(self):
        self.cards.sort()
```

````python
deck = Deck(Deck.make_cards())
deck.shuffle()

print("Before sort:")
for card in deck.cards[:5]:
    print(card)

deck.sort()

print("\nAfter sort:")
for card in deck.cards[:5]:
    print(card)
````

## **Parents and children (Inheritance)**
Inheritance lets you define a new class based on an existing one, so it reuses methods and attributes

````python
class Hand(Deck):
    """Represents a hand of playing cards."""
    def __init__(self, label=''):
        self.label = label
        self.cards = []

hand = Hand("Player 1")
hand.put_card("Ace of Spades")
hand.put_card("King of Hearts")

print(hand.cards)
````
