# Conditionals and Recursion

## Integer Division and Modulus
Python supports integer division (`//`) which gives the quotient with the remainder discarded, and the modulus operator `%` which gives the remainder of a division. For example, `x % 10` returns the rightmost digit of `x`.

## Boolean Expressions
A boolean expression is an expression that evaluates to `True` or `False`. Python uses relational operators like `==`, `!=`, `>`, `<`, `>=`, <`=` to compare values.

## Logical Operators
Logical operators `and`, `or`, and `not` combine boolean expressions. `x > 0 and x < 10` is `True` only if both conditions hold.

## Statements
The `if` statement lets your program make decisions. If the condition is true, the indented block runs.

## `else` Clause
`if` can be followed by `else` to specify an alternate block when the condition is false.

## Chained Conditionals (`elif`)
When there are multiple possibilities, you can use `elif` (“else if”) between `if` and `else`. Python evaluates each condition in order and executes only the first true block.

## Recursion
A recursive function calls itself. To avoid infinite recursion, each recursive function must have a base case, a condition under which it does not recurse.

## Infinite Recursion
If recursion never reaches a base case, it continues indefinitely until Python throws a `RecursionError`.

## Keyboard Input
Use `input()` to read text from the user. It returns a string. To convert to an integer, wrap it in `int()`.

## Debugging
Error messages show what kind of error and where it occurred. For runtime errors, the cause might be earlier than the line indicated.

```python
import time

# the total number of seconds since the Unix epoch until now.
total_seconds = int(time.time())

# the total number of seconds in a day
seconds_per_day = 24 * 60 * 60

# the remaining number of seconds in the current day
days = total_seconds // seconds_per_day

# the remaining number of seconds in the current day
remaining_seconds = total_seconds % seconds_per_day

# current time
hours = remaining_seconds // 3600
remaining_seconds = remaining_seconds % 3600

minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print("Days since Jan 1, 1970:", days)
print("Current time (UTC):", hours, ":", minutes, ":", seconds)
```
