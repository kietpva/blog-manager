## Loops and strings
This section explains how `for` loops work with strings. In Python, a string is a sequence of characters, and a `for` loop can iterate over each character in a string. Each iteration assigns the next character to the loop variable.

A common pattern is to use a `for` loop with a string to analyze or build a new string. Because strings are immutable, you often accumulate a result in a new string or a list.

Examples illustrate looping through all characters in a string, counting occurrences of a specific character, and creating a new string with modified content.

**Example**

```python
word = "banana"
count = 0

for letter in word:
    if letter == "a":
        count = count + 1

print(count)  # 3
```

This counts how many times `a` appears in `banana`

**Example loop to build a new string**

```python
word = "think"
result = ""

for letter in word:
    result = result + letter + "."
print(result)  # t.h.i.n.k.
```

## Reading the word list

This section explains how to read a list of words from a file and store them in a Python list for further processing, such as checking whether a word is in the list or analyzing patterns.

The example uses a file called `words.txt`, which contains one word per line. To read the file:

1. Open the file using `open()`.
2. Loop through each line in the file with a `for` loop.
3. Strip off the newline (`\n`) at the end of each line using `strip()` or `rstrip()` before adding the word to a list.

After reading, you get a list called `wordlist` that contains all the words. This list can then be passed to other functions for analysis.

The section also introduces a helper function (often called `read_words()` or similar) to encapsulate this logic and return the list of words.

**Example:**

```python
def read_words(filename):
    """Return a list of words from a file."""
    wordlist = []
    fin = open(filename)
    for line in fin:
        word = line.strip()  # remove newline
        wordlist.append(word)
    return wordlist

words = read_words('words.txt')
print(words[:10])  # show first 10 words
```

## Updating variables

In Python, you often need to change (update) the value stored in a variable. This section explains how to update variables using assignment and simple arithmetic operations.

The most common pattern is to increment a variable inside a loop. For example:

```python
count = 0
for i in range(10):
    count = count + 1
```

Each time through the loop, `count` gets increased by 1. This is called an accumulator pattern, where you accumulate a result as you iterate.

Python provides a shorthand for updating variables using augmented assignment operators such as `+=`, `-=`, `*=`, etc. So the increment above can be written as:

```python
count += 1
```

This section also explains how to update variables in more complex ways, such as building up a string by concatenation:

```python
result = ''
for ch in word:
    result = result + ch + '.'
```

or accumulating a sum. These patterns are fundamental to working with loops.
