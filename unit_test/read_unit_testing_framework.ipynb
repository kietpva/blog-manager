# **Unit testing framework**
## **Basic example**
**Purpose**
This example demonstrates the basic usage of `unittest`
- Defining test cases
- Using assertions
- Runing tests

`````python
import unittest


class TestStringMethods(unittest.TestCase):
    def test_upper(self):
        self.assertEqual("foo".upper(), "FOO")

    def test_isupper(self):
        self.assertTrue("FOO".isupper())
        self.assertFalse("Foo".isupper())

    def test_split(self):
        s = "hello world"
        self.assertEqual(s.split(), ["hello", "world"])
        with self.assertRaises(TypeError):
            s.split(2)

# 
# if __name__ == "__main__":
#     unittest.main()

# argv=[...] -> ignore Jupyter arguments
# exit=False -> prevents kernel shutdown
unittest.main(argv=["first-arg-is-ignored"], exit=False)
`````

1. **import unittest**
   Import the unittest module to access the testing framework.
   ```python
   import unittest
   ```
2. **Create test case**
   - The class must inherit from `unittest.TestCase`
   - Any method starting with `test_` is treated as an individual test
   ```python
   class TestStringMethods(unittest.TestCase):
   ```
3. **Case 1: `test_upper`**
   Checks whether the actual result equals the expected value
   ```python
   self.assertEqual('foo'.upper(), 'FOO')
   ````
4. **Case 2: `test_isupper`**
   ```python
   self.assertTrue('Foo'.isupper())
   self.assertFalse('Foo'.isupper())
   ```
   - `assertTrue`: expects the expression to be `True`
   - `assertFalse`: expects the expression to be `False`
   
5. **Case 3: `test_split` + Exception**
   ```python
   with self.assertRaises(TypeError):
    s.split(2)
   ```
   Verifies that the code raises the expected exception type.
6. **Run test**
   ```python
   if __name__ == '__main__':
    unittest.main()
   ```

## **Command-Line Interface**
### **Run test with commandline**
You can run tests from the command line with `unittest` for modules, class, or individual test methods

- Run all tests in the `test_module1` and `test_module2` modules
  ```bash
  python -m unittest test_module1 test_module2
  ```

- Run all test in a specific class
  ```bash
  python -m unittest test_module.TestClass.test_method
  ```
- Run only test method
  ```bash
  python -m unittest test_module.TestClass.test_method
  ```
### **Run tests by file path**
```bash
python -m unittest tests/test_example.py
```
- You can specify a test file by its file path
- The file must be importable as a Python module

### **Test dicovery**
```bash
python -m unittest discover
```
- `unittest` automatically discovers tests based on conventions
- The default convention is files starting with `test*.py`
- Running `python -m unittest` is equivalent to `discover`

### **Command-line options**

| Command-line options |   Function   |
| -------------------- | ------------ |
| `-v`/`--verbose`     | Show more detailed test output |
| `-f`/`--failfast`    | Stops immediately on the first falure or error |
| `-b`/`--buffer`      | Buffers stdout and stderr during test execution |
| `c`/`--catch`        | Catches `ctrl+c` and exits cleanly |
| `--locals`           | Shows local variables in tracebacks |
| `-k PATTERN`         | Run only tests whose names match the pattern |

### **Show help**
```bash
python -m unittest -h
```

Displays all available command-line options

## **Test Discovery**
*Test discovery* is a mechanism that lets `unittest` automatically find all tests in your project and run them without specifying each one manually

### **Enable test discovery**
```bash
cd project_directory
python -m unittest discover
```

change into your project directory, then run `python -m unittest discover` to automatically find tests

### **The discover sub-command has the following options**
| options |   Function   |
| -------------------- | ------------ |
| `-v`/`--verbose`     | Show more detailed test output |
| `-f`/`--failfast`    | Stops immediately on the first falure or error |
| `-b`/`--buffer`      | Buffers stdout and stderr during test execution |
| `c`/`--catch`        | Catches `ctrl+c` and exits cleanly |
| `--locals`           | Shows local variables in tracebacks |
| `-k PATTERN`         | Run only tests whose names match the pattern |

### **Test file naming convention**
- Test files must be importable Python modules (valid file names)
- Discovery finds test files based on the file name pattern(`test*.py`)

### **How it works**
- `TestLoader.discover()` walks directories from the `start-directory` using the pattern
- It imports each test file and adds tests to a `TestSuite`

## **Organizing test code**

### **Basic test case**
A test case is the smallest unit of testing, it sets up a scenario and checks correctness. In `unittest`, each test case is represented by a class inheriting from `unittest.TestCase`
```python
import unittest

class DefaultWidgetSizeTestCase(unittest.TestCase):
    def test_default_widget_size(self):
        widget = Widget('The widget')
        self.assertEqua(widget.size(), (50, 50))
```
Use `assert*` methods to check for expected outcomes

### **setUp and tearDown**
if many tests share setup code, put it in `setUp()`, which runs before each test method
```python
class WidgetTestCase(unittest.TestCase):
    def setUp(self):
        self.widget = Widget('The widget')
```
Likewise use `teaDown()` to clean up after each test

### **Group tests**
You can group tests into a TestSuite for grouped excution
```python
def suite():
    suite = unittest.TestSuite()
    suite.addTest(WidgetTestCase('test_default_widget_size'))
    suite.addTest(WidgetTestCase('test_widget_resize'))

    return suite
```

### **Separate test modules**
You can put tests in the same file as the code, but it's recommended to put them in a separate module like `test_widget.py`
**Benefits:**
- Run tests independently
- Easy to separate from shopped code
- Less temptation to change tests for wrong reasons
- Easier refactor of code under test

### **Re-using old test code**
If you have existing test functions not in `TestCase` from, `unittest` provides `FunctionTestCase` to wrap them
```python
testcase = unittest.FunctionTestCase(testSomething, setup=makeSomethingDB, testDown=deleteSomethingDB)
```
However, using `FunctionTestCase` is not recommended long-term, better write proper `TestCase` classes

## **Re-using old test code**
This section explains how to reuse old test functions that are not inside a `TestCase` class

### **FunctionTestCase**
If you have legacy test functions (not in classes), `unittest` provides `FunctionTestCase` to wrap them into proper test cases the framework can run
```python
testcase = unittest.FunctionTestCase(testSomething, setup=makeSomethingDB, testDown=deleteSomethingDB)
```
In this snippet
- `testSomething` is the old test function
- `setUp` is an optional setup helper
- `teaDown` is an optional cleanup helper

### **Run the converted tests using a TestSuite**
You can add this `FunctionTestCase` to a `TestSuite` to run it alongsid other tests
```python
suite = unittest.TestSuite()
suite.addTest(testcase)
```
Then use a test runner to execute the whole suite
### 
Although `FunctionTestCase` is useful for quick reuse, Python recommends rewriting tests as `TestCase` class for maintainability

## **Skipping tests**
`unittest` lets you skip an individual test method or an entire test class when you don't want it executed

`@unittest.skip(reason)`: Skip a test unconditionally, `reason` explains why it’s skipped
```python
@unittest.skip("demonstrating skipping")
def test_nothing(self):
    ...
```

`@unittest.skipIf(condition, reason)`: Skip a test if the condition is True

`@unittest.skipUnless(condition, reason)`: Skip a test unless the condition is True

`TestCase.skipTest(reason)` and `SkipTest`: You can call inside `setUp()` or a test to skip it when something is unavailable

**Effects of skipped tests**
Skipped tests don't run `setup()` or `tearDown()`, and the result is recorded in `skipped` with the reason

**Expected failures**
Sometimes a test is expected to fail (due to a known issue) but shouldn't count as a real failure in results.

`@unittest.expectedFailure`
Use this decorator to mark a test as an expected failure
```python
class expectedFailureTestCase(unittest.TestCase):
    @unittest.expectedFailure
    def test_fail(self):
        self.assertEqual(1, 0, 'broken')
```
if the test acually fails/errors, it is considered expected(not counted as a failure)
if the test passes, it is recorded as an `unexpectedSuccess`

## **Distinguishing test iterations using subtests**
- when you have multiple tests that differ only by small parameter values, you can use subtests to run them inside a single test method
- Subtests allow you to report each case separately, and even if one fails, the others still run

### **Using `subTest()`**
- use `with self.subTest(...)` inside a test method
- you can pass a message and keyword args to help identify each subtest in output

```python
class NumbersTest(unittest.TestCase):
    def test_even(self):
        for i in range(0, 6):
            with self.subTest(i-i):
                self.assertEqua(i% 2, 0)
```
The snippet tests values from 0 to 5, with each value treated as a separate subtest

### **Output when a subtest fails**
If a subtest fails (e.g., when `i=1`), the output reports the specific test method + parameter that failed

```makefile
FAIL: test_even (__main__.NumbersTest.test_even) (i=1)
```
This helps you see exactly which case failed without stopping the whole test method

### **When to use subtests**
- when you repeat the same test logic with different parameters
- when you want individual reporting per case instead of stopping early

## **Classes and functions**
This section describes the key classes and functions that `unittest` provides for building, grouping, running, and reporting tests

### **Test case**
`TestCase` represents a single test. Each method starting with `test` is treated as an individual test
- `run(result=None)`: runs the test and records the results in a `TestResult`
- `assertX()` methods: assertion methods (e.g. `equal`, `true`, `rauses`, etc...)
- Additional methods such as `addCleanup`, `enterContext`, etc are provided for resource management

### **Assert methods**
`TestCase` includes many assert methods to check values, types, exceptions, logs, etc.
- `assertEqua(a, b)`: a == b
- `assertTrue(x)` / `assertFalse(x)`: check true/false
- `assertRaises(exc)`: check exception
- `assertLogs(logger, level)`: check logging

There are many more for almost every common test scenario

### **TestSuite**
`TestSuite` is a class to group multiple test cases or suites together
```python
suite = unittest.TestSuite()
suite.addTest(testcase)
```
A TestSuite behaves like a test case but doesn't check anything itself, it just contains multiple tests

### **TestLoader**
`TestLoader` is used to create TestSuites from TestCase classes or modules.
- `loadTestsFromTestCase(testCaseClass)`: loads tests from a test case class
- `testMethodPrefix`: the method name prefix used to identify testMethods.
- `testNamePatterns`: patterns used to filter test names
You can customize how the loader finds and selects tests

### **TestResult**
`TestResult` holds information about tests run: failures, errors, skipped tests, unexpected successes, etc.
- `errors`: tests that raise unexpected exceptions
- `failures`: tests that fail assertions
- `skipped`: tests that are skipped
- `expectedFailure`/`unexpectedSuccess`: Tests that are expected to fail, and tests that unexpectedly succeed
Test runners and other tools usually use `TestResult` to report results

### **TextTestRunner**
`TextTestRunner` is the most common runner, printing rtest results to the terminal
```python
runner unittest.TextTestRunner()
runner.run(suite)
```
You can adjust verbosity, buffering, failfast behavior, etc...

### **unittest.main()**
`unittest.main()` is a convenience function to run tests in a script
```python
if __name__ == '__main))':
    unittest.main()
```
It automatically uses the `TestLoader` to find tests and the `TextTestRunner` to run them

### **Special test classes (async)**
Python also provides `IsolatedAsyncioTestCase` to test asynchronous code

## **Class and Module Fixtures**
This section explains how `unittest` handles class-level and module-level fixtures to setup and tear down resources before/after groups of tests

### **How it works**
Class and module level fixtures are implemented by `TestSuite`. When running tests, if a new test class is encountered
- `tearDownClass()` of the previous class (if any) will be called
- `setUpClass()` of the new class will be called

This lets you set up state once for all tests in that class rather than before each test

### **Module fixtures**
Similarly, if the next test is in a different module than the previous one:
- `tearDownModule()` of the previous module will be called
- `setUpModule()` of the new module will be called
This allows setup/cleanup before and after all tests in a module file

### **After all tests**
After all tests hae completed
- Final fixtures such as `tearDownClass()` and `tearDownModule()` are executed

This ensures that all shared resources are cleaned up

## **Signal Handling**
This section explains how `unittest` handles the **Ctrl-C** signal(SIGINT) during a test run

### **Purpose**
When running tests from the command line and pressing **Ctrl-C**, `unittest` provides a "frendly" way to handle stopping the tests

### `-c/--catch` and `catchbreak`
The `-c/--catch` command-line option or the `catchbreak` parrameter to `unittest.main()` allows:
- On the first Ctrl-C, let the current test finish and then end the test run
- On the second Ctrl-C, raise a normal

### **Integration with existing handlers**
If the code under test has installed its own handler for `signal.SIGINT`, `unittest` respects that existing handler and will invoke the default handler only when appropriate

this ensures compatibility with code that custom-handles Ctrl-C

### Utility functions
`unittest` offers several utility functions for advanced test frameworks:
1. `unittest.installHandler()`
   - install the Ctrl-C handler for `unittest`
   - When a SIGINT (Ctrl-C) is received, all registered TestResult objects will receive a stop() call
2. `unittest.registerResult(result)`
   - Register a `TestResult` object for Ctrl-C handling
   - The registration uses weak references, so it does not prevent the objects from being garbage collected
3. `unittest.removeResult(result)`
   - Remove a previously registered `TestResult`
   - After being removed, `stop()` will no longer be called on this result when Ctrl-C occurs.
4. `unittest.removeHandler(function=None)`
   - Calling with no arguments → removes the installed Ctrl-C handler.
   - It can also be used as a decorator to temporarily disable the handler during a specific test.
```python
@unittest.removeHandler
def test_signal_handling(self):
    ...
```
