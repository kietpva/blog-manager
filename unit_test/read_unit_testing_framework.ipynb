# **Unit testing framework**
## **Basic example**
**Purpose**
This example demonstrates the basic usage of `unittest`
- Defining test cases
- Using assertions
- Runing tests

`````python
import unittest


class TestStringMethods(unittest.TestCase):
    def test_upper(self):
        self.assertEqual("foo".upper(), "FOO")

    def test_isupper(self):
        self.assertTrue("FOO".isupper())
        self.assertFalse("Foo".isupper())

    def test_split(self):
        s = "hello world"
        self.assertEqual(s.split(), ["hello", "world"])
        with self.assertRaises(TypeError):
            s.split(2)

# 
# if __name__ == "__main__":
#     unittest.main()

# argv=[...] -> ignore Jupyter arguments
# exit=False -> prevents kernel shutdown
unittest.main(argv=["first-arg-is-ignored"], exit=False)
`````

1. **import unittest**
   Import the unittest module to access the testing framework.
   ```python
   import unittest
   ```
2. **Create test case**
   - The class must inherit from `unittest.TestCase`
   - Any method starting with `test_` is treated as an individual test
   ```python
   class TestStringMethods(unittest.TestCase):
   ```
3. **Case 1: `test_upper`**
   Checks whether the actual result equals the expected value
   ```python
   self.assertEqual('foo'.upper(), 'FOO')
   ````
4. **Case 2: `test_isupper`**
   ```python
   self.assertTrue('Foo'.isupper())
   self.assertFalse('Foo'.isupper())
   ```
   - `assertTrue`: expects the expression to be `True`
   - `assertFalse`: expects the expression to be `False`
   
5. **Case 3: `test_split` + Exception**
   ```python
   with self.assertRaises(TypeError):
    s.split(2)
   ```
   Verifies that the code raises the expected exception type.
6. **Run test**
   ```python
   if __name__ == '__main__':
    unittest.main()
   ```

## **Command-Line Interface**
### **Run test with commandline**
You can run tests from the command line with `unittest` for modules, class, or individual test methods

- Run all tests in the `test_module1` and `test_module2` modules
  ```bash
  python -m unittest test_module1 test_module2
  ```

- Run all test in a specific class
  ```bash
  python -m unittest test_module.TestClass.test_method
  ```
- Run only test method
  ```bash
  python -m unittest test_module.TestClass.test_method
  ```
### **Run tests by file path**
```bash
python -m unittest tests/test_example.py
```
- You can specify a test file by its file path
- The file must be importable as a Python module

### **Test dicovery**
```bash
python -m unittest discover
```
- `unittest` automatically discovers tests based on conventions
- The default convention is files starting with `test*.py`
- Running `python -m unittest` is equivalent to `discover`

### **Command-line options**

| Command-line options |   Function   |
| -------------------- | ------------ |
| `-v`/`--verbose`     | Show more detailed test output |
| `-f`/`--failfast`    | Stops immediately on the first falure or error |
| `-b`/`--buffer`      | Buffers stdout and stderr during test execution |
| `c`/`--catch`        | Catches `ctrl+c` and exits cleanly |
| `--locals`           | Shows local variables in tracebacks |
| `-k PATTERN`         | Run only tests whose names match the pattern |

### **Show help**
```bash
python -m unittest -h
```

Displays all available command-line options

## **Test Discovery**
*Test discovery* is a mechanism that lets `unittest` automatically find all tests in your project and run them without specifying each one manually

### **Enable test discovery**
```bash
cd project_directory
python -m unittest discover
```

change into your project directory, then run `python -m unittest discover` to automatically find tests

### **The discover sub-command has the following options**
| options |   Function   |
| -------------------- | ------------ |
| `-v`/`--verbose`     | Show more detailed test output |
| `-f`/`--failfast`    | Stops immediately on the first falure or error |
| `-b`/`--buffer`      | Buffers stdout and stderr during test execution |
| `c`/`--catch`        | Catches `ctrl+c` and exits cleanly |
| `--locals`           | Shows local variables in tracebacks |
| `-k PATTERN`         | Run only tests whose names match the pattern |

### **Test file naming convention**
- Test files must be importable Python modules (valid file names)
- Discovery finds test files based on the file name pattern(`test*.py`)

### **How it works**
- `TestLoader.discover()` walks directories from the `start-directory` using the pattern
- It imports each test file and adds tests to a `TestSuite`

## **Organizing test code**

### **Basic test case**
A test case is the smallest unit of testing, it sets up a scenario and checks correctness. In `unittest`, each test case is represented by a class inheriting from `unittest.TestCase`
```python
import unittest

class DefaultWidgetSizeTestCase(unittest.TestCase):
    def test_default_widget_size(self):
        widget = Widget('The widget')
        self.assertEqua(widget.size(), (50, 50))
```
Use `assert*` methods to check for expected outcomes

### **setUp and tearDown**
if many tests share setup code, put it in `setUp()`, which runs before each test method
```python
class WidgetTestCase(unittest.TestCase):
    def setUp(self):
        self.widget = Widget('The widget')
```
Likewise use `teaDown()` to clean up after each test

### **Group tests**
You can group tests into a TestSuite for grouped excution
```python
def suite():
    suite = unittest.TestSuite()
    suite.addTest(WidgetTestCase('test_default_widget_size'))
    suite.addTest(WidgetTestCase('test_widget_resize'))

    return suite
```

### **Separate test modules**
You can put tests in the same file as the code, but it's recommended to put them in a separate module like `test_widget.py`
**Benefits:**
- Run tests independently
- Easy to separate from shopped code
- Less temptation to change tests for wrong reasons
- Easier refactor of code under test

### **Re-using old test code**
If you have existing test functions not in `TestCase` from, `unittest` provides `FunctionTestCase` to wrap them
```python
testcase = unittest.FunctionTestCase(testSomething, setup=makeSomethingDB, testDown=deleteSomethingDB)
```
However, using `FunctionTestCase` is not recommended long-term, better write proper `TestCase` classes

## **Re-using old test code**
This section explains how to reuse old test functions that are not inside a `TestCase` class

### **FunctionTestCase**
If you have legacy test functions (not in classes), `unittest` provides `FunctionTestCase` to wrap them into proper test cases the framework can run
```python
testcase = unittest.FunctionTestCase(testSomething, setup=makeSomethingDB, testDown=deleteSomethingDB)
```
In this snippet
- `testSomething` is the old test function
- `setUp` is an optional setup helper
- `teaDown` is an optional cleanup helper

### **Run the converted tests using a TestSuite**
You can add this `FunctionTestCase` to a `TestSuite` to run it alongsid other tests
```python
suite = unittest.TestSuite()
suite.addTest(testcase)
```
Then use a test runner to execute the whole suite
### 
Although `FunctionTestCase` is useful for quick reuse, Python recommends rewriting tests as `TestCase` class for maintainability
