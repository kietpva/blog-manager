# **Python's unittest: Writing Unit Tests for Your Code**
## **Testing Your Python Code**
This section introduces why and how to write tests for your Python code using the `unittest` framework

### **Code Testing**
- Code testing(software testing) is a fundamental part of the modern software development cycle
- Through testing, you verify that software works as expected and meets its requirements
- Testing enforces code quality and robustness

### **When and why testing happens**
- You perform testing during the development stage of an application or project
- Tests are written to isolate sections of your code and verify their correctness
- A well-written test suite can also serve as documentation for the project

### **Unit testing concept**
- There are many testing concepts and techniques, but unit testing is an important and relevant one
- A unit test is a test that operates on an individual unit of software
- The goal of a unit test is to validate that the tested unit works as designed

### **What is a unit?**
- A unit is often a small part of a program that takes some inputs and produces and output
- Functions, methods, and other callables are typical examples of units to test

### **Unit testing tools in Python**
- Python provides several tools to write, origanize, run, and automate unit tests
- In Python's standard library, there are two main unit testing tools:
  - doctest
  - unittest

### **doctest**
- `doctest` is a lightweight and straightforward testing framework
- It reads test cases directly from documentation and code docstrings
- `doctest` ships ưith Python a pat o it batteries-included philosophy

### **unittest**
- `unittest` is also a testing framework, but more complete and powerful than doctest
- `unittest` allows you to build structured, clear, and extensible unit tests
- The following sections forcus on using `unittest` to create suitable unit tests for Python code

## **Getting to Know Python’s unittest**
- This section introduces `unittest`, Python's built-in unit testing framework that you can use without installing extra packages
- `unittest` uses an object-oriented approach and is inspired by JUnit(a testing framework for Java)

**Key concepts**
- a TestCase is an individual unit of testing that checks output for a given set of inputs
- A testSuite is a collection of TestCase(and/or other suites) that are executed as a group
- A test fixture is the setup and teardown logic to prepare and clean up before and after tests
- A test runner is the component that excutes tests and reports results

## **Origanizing Your Tests With the `TestCase` Class**
- To write unit tests with `unittest`, you create a class that inherits from `unittest.TestCase`
- Within the class, each method whose name starts with `test_` is treated as an individual test
### **Writing your first test case**
For example, to test the built-in `abs()` function, you can write:

````python
import unittest

class TestAbsFunction(unittest.TestCase):
    def test_positive_number(self):
        self.assertEqual(abs(10), 10)

    def test_negative_number(self):
        self.assertEqual(abs(-10), 10)

    def test_zero(self):
        self.assertEqual(abs(0), 0)
````

Each test methods checks a specific input and its expected output

### **Creating test cases for custom functions**
For a custom function `categorize_by_age()`, you also create a test class inheriting `TestCase`
```python
class TestCategorizeByAge(unittest.TestCase):
    def test_child(self):
        self.assertEqual(categorize_by_age(5), "Child")
    
    def test_adolescent(self):
        self.assertEqual(categorize_by_age(15), "Adolescent")
    
    def test_adult(self):
        self.assertEqual(categorize_by_age(30), "Adult")
```
Naming test classes with `Test` is a common convention that helps readability and test discovery

### **Test discovery and naming**
- By default, `unittest` uses the filename pattern `test*.py` to automatically discover and run test modules
- For example, name the test module `test_age.py` lets unittest automatically detrct and run it

### **using multiple assertions**
- You can include multiple assert statements in a single test method to reduce repetitive code
- For example: checking boundary values between age ranges:
```python
def test_boundary_child_adolescent(self):
    self.assertEqual(categorize_by_age(9), "child")
    self.assertEqual(categorize_by_age(10), "Adolescent)
```

## **Exploring the Available Assert Methods**
This section lists and explains the assertion methods that `unittest.TestCase` provides for checking various conditions in unit tests

`unittest.TestCase` provides more than 20 different assertion methods to cover a wide range of testing needs

### **Basic Comparision Assertions**
| method | meaning |
| ------ | ------------ |
| assertEqual(a, b) | Check that `a == b` |
| assertNotEqual(a, b) | Check that `a != b` |
| assertTrue(x) | check that `x` is `True` |
| assertFalse(x) | Check that `x` is `False` |

### **Identity Assertions**
| method | meaning |
| ------ | ------- |
| assertIs(a, b) | Check `a is b` |
| assertIsNot(a, b) | Check `a is not b` |
| assertIsNone(x) | Check `x is None` |
| assertIsNotNone (x) | Check `x is not None` |

### **Squences & Collections**
| method | meaning |
| ------ | ------- |
| assertListEqual(a, b) | compare two lists |
| assertTupleEqual(a, b) | compare two tuples |
| assertSetEqual(a, b) | compare two sets |
| assertDictEqua(a, b) | compare two dicts |
| assertMultiLineEqua(a, b) | compare multi-line strings |

### **Membership Tests**
| method | meaning |
| ------ | ------- |
| AssertIn(a, b) | Check `a in b` |
| AssertNotIn(a, b) | Check `a not in b` |

### **Type & instance**
| method | meaning |
| ------ | ------- |
| assertIsInstance(a, b) | Check `isinstance(a, b)` |
| assertNotIsInstance(a, b) | Check **NOT** instance |

### **Exception * warning**
| method | meaning |
| ------ | ------- |
| assertRaise(E, func) | Expect exception |
| assertRaisesRegex(E, r, func) | Exception message matches regex |
| assertWarns(W, func) | Expect warning |
| assertWarnsRegex(W, r, func) | Warning matches regex |

### **Logging**
| method | meaning |
| ------ | ------- |
| assertLogs(logger, level) | expect logs |
| assertNoLogs(logger, level) | expect no logs |

## **Grouping Your Tests with the TestSuite Class**
This section explains how to use `unittest` `TestSuite` class to group multiple tests together and execute them as a group

Grouping tests is useful in larger projects or when you want to run specific subsets of tests selectively

### **Benefits of using TestSuite**
- Organize tests into logical groups(e.g., airthmetic vs.statistical)
- Run tests by level (unit, integration, system)
- Run enviroment-specific tests(Windows, Linux, etc.)

### **Creating a TestSuite**
#### **Using the constructor with a list of tests**
**Method 1: Using the constructor with a list of tests**
```python
def make_suite():
    arithmetic_tests = [
        TestArithmeticOperations("test_add"),
        TestArithmeticOperations("test_subtract"),
        TestArithmeticOperations("test_multiply"),
        TestArithmeticOperations("test_divide"),
    ]
    return unittest.TestSuite(tests=arithmetic_tests)
```
- Create a list of specific test methods
- Pass the list into `TestSuite()`

**Method 2: Use `.addTest()`**
```python
arithmetic_suite = unittest.TestSuite()
arithmetic_suite.addTest(TestArithmeticOperations("test_add"))
arithmetic_suite.addTest(TestArithmeticOperations("test_subtract"))
...
```
Add each test to the suite with `.addTest()`

**Method 3: Use `.addTests()`**
```python
statistical_suite = unittest.TestSuite()
statistical_suite.addTests([
    TestStatisticalOperations("test_mean"),
    TestStatisticalOperations("test_median_odd"),
    ...
])
```
- `.addTests()` takes an iterable of multiple tests
- Useful to add batches of tests at once

### **TestSuite Run**
```python
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
```
use `TextTestRunner` to execute the suite and print results

### **Running tests by group**
By creating multiple suites(like arithmetic and statistical), you can run each group of tests separately

### **Integration with `load_tests()`**
You can deine a `load_tets()` function in your module to automatically build a test suite using the loader
```python
def load_tests(loader, standard_tests, pattern):
    suite = unittest.TestSuite()
    suite.addTests(loader.loadTestsFromTestCase(TestArithmeticOperations))
    suite.addTests(loader.loadTestsFromTestCase(TestStatisticalOperations))
    return suite
```

## **Creating Test Fixture**

### **What a fixture**
A test fixture is the data, connection, or state that multiple test methods share

Fix tures help you reuse setup/cleanup logic and avoid repeating code in each test

### **setUp() and tearDown()**
**setUp(self): Run before each test**
The `setUp()` method is called before every test method in a `TestCase` to establish fixtures

**tearDown(self): Run after each test**
The `tearDown()` method runs after each test method to clean up what was set up

### **Other fixture levels (besides setUp / tearDown)**
You can also define module-level fixture that run before/after the entire module's tests:
- `setUpModle()`
- `tearDownModule()`
Module-level fixtures run once for all tests in a module

### **Purpose of fixtures**
- Reduce repeated setup/cleanup code
- Ensure each test gets a consistent environment
- Separate test logic from fixture setup

## **Debugging Failing Tests**
when writing unit tests, sometimes a test doesn't pass

This section shows how to analyze and fix failing tests

### **Understanding test output**
when running `unittest`, the output typically shows:
- `.`: test passed
- `F`: test failed
- E: test error(unexpected exception)

### **Traceback details**
when a test fails or errors, `unittest` prints detailed traceback showing:
- where the failure occurred
- Code line that caused it
- Actual vs expected values

This is the first source of information you use to debug

### **Failfast**
You can run tests with `failfast` so it stops at the first failure

**example CLI**
```bash
python -m unittest -f
```

**or programmatically**
```python
unittest.TextTestRunner(failfast=True).run(suite)
```

### **Using a debugger**
You can place a brakpoint in the test or in the code under test

In terminal, you can use
```bash
python -m pdb test_example.py
```
or in you IDE(VS Code/PyCharm), run tests in debug mode to step through code line by line

### **View detailed assertion output**
some assert methods print detailed info on failure:
- `.assertEqual(a, b)` shows both actual and expected values
You can add a custom message(`msg=`) to an assert to clarify the intention when it fails

### **Increase verbosity to read the output more clearly**
Use `verbosity=2` to display each test's name and more details instead of just dots
CLI example:
```bash
python -m unittest -v
```
