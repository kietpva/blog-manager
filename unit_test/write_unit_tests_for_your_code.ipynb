# **Python's unittest: Writing Unit Tests for Your Code**
## **Testing Your Python Code**
This section introduces why and how to write tests for your Python code using the `unittest` framework

### **Code Testing**
- Code testing(software testing) is a fundamental part of the modern software development cycle
- Through testing, you verify that software works as expected and meets its requirements
- Testing enforces code quality and robustness

### **When and why testing happens**
- You perform testing during the development stage of an application or project
- Tests are written to isolate sections of your code and verify their correctness
- A well-written test suite can also serve as documentation for the project

### **Unit testing concept**
- There are many testing concepts and techniques, but unit testing is an important and relevant one
- A unit test is a test that operates on an individual unit of software
- The goal of a unit test is to validate that the tested unit works as designed

### **What is a unit?**
- A unit is often a small part of a program that takes some inputs and produces and output
- Functions, methods, and other callables are typical examples of units to test

### **Unit testing tools in Python**
- Python provides several tools to write, origanize, run, and automate unit tests
- In Python's standard library, there are two main unit testing tools:
  - doctest
  - unittest

### **doctest**
- `doctest` is a lightweight and straightforward testing framework
- It reads test cases directly from documentation and code docstrings
- `doctest` ships ưith Python a pat o it batteries-included philosophy

### **unittest**
- `unittest` is also a testing framework, but more complete and powerful than doctest
- `unittest` allows you to build structured, clear, and extensible unit tests
- The following sections forcus on using `unittest` to create suitable unit tests for Python code

## **Getting to Know Python’s unittest**
- This section introduces `unittest`, Python's built-in unit testing framework that you can use without installing extra packages
- `unittest` uses an object-oriented approach and is inspired by JUnit(a testing framework for Java)

**Key concepts**
- a TestCase is an individual unit of testing that checks output for a given set of inputs
- A testSuite is a collection of TestCase(and/or other suites) that are executed as a group
- A test fixture is the setup and teardown logic to prepare and clean up before and after tests
- A test runner is the component that excutes tests and reports results

## **Origanizing Your Tests With the `TestCase` Class**
- To write unit tests with `unittest`, you create a class that inherits from `unittest.TestCase`
- Within the class, each method whose name starts with `test_` is treated as an individual test
### **Writing your first test case**
For example, to test the built-in `abs()` function, you can write:

````python
import unittest

class TestAbsFunction(unittest.TestCase):
    def test_positive_number(self):
        self.assertEqual(abs(10), 10)

    def test_negative_number(self):
        self.assertEqual(abs(-10), 10)

    def test_zero(self):
        self.assertEqual(abs(0), 0)
````

Each test methods checks a specific input and its expected output

### **Creating test cases for custom functions**
For a custom function `categorize_by_age()`, you also create a test class inheriting `TestCase`
```python
class TestCategorizeByAge(unittest.TestCase):
    def test_child(self):
        self.assertEqual(categorize_by_age(5), "Child")
    
    def test_adolescent(self):
        self.assertEqual(categorize_by_age(15), "Adolescent")
    
    def test_adult(self):
        self.assertEqual(categorize_by_age(30), "Adult")
```
Naming test classes with `Test` is a common convention that helps readability and test discovery

### **Test discovery and naming**
- By default, `unittest` uses the filename pattern `test*.py` to automatically discover and run test modules
- For example, name the test module `test_age.py` lets unittest automatically detrct and run it
