# **Typing**
## **Basic Types**
Basic types are built-in Python types used in annotations to:
- Describe function parameters
- Describe return values
- Improve readability and tooling support

```python
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

print(calculate_total(price=20.2, quantity=2))
print(type(calculate_total(price=20.2, quantity=2)))
```

## **Collections(List, Dict, Set)**
Collections let you specify what type each element contains, not just the container itself

```python
from typing import List, Dict

score: List[int] = [90, 85, 80]

users: Dict[int, str] = {
    1: 'Alice',
    2: 'Bob'
}
```

- `int` is key
- `str` is value

## **Generics**
Generics are used when:
- Code should work with many types
- But stull preserve type relationships

**Generic Function**

```python
from typing import TypeVar

T = TypeVar("T")


def get_first(items: list[T]) -> T:
    return items[0]
```

**Generic Class**

```python
from typing import Generic, TypeVar

T = TypeVar("T")


class Box(Generic[T]):
    def __init__(self, value: T):
        self.value = value

    def get(self) -> T:
        return self.value


box = Box
text_box = Box[str]("hello")
```

## **Callables**
`Callable` is used when functions are passed as values and must follow a specific signature

```python
from typing import Callable


def calculate(
    x: int,
    y: int,
    operation: Callable[[int, int], int]
) -> int:
    return operation(x, y)

def add(a: int, b: int) -> int:
    return a + b

calculate(3, 4, add)
```

## **Union & Optional**
- `Union[A, B]`: value can be one of several types
- `Optional[T]`: shorthand for `Union[T, None]`

```python
from typing import Optional

def find_user(user_id: int) -> Optional[str]:
    if user_id == 1:
        return "Alice"
    return None
```

## **Literal Types**
`Literal` restricts a value to a specific set of constant values
- status
- role
- mode
- flag

```python
from typing import Literal

Role = Literal["admin", "user", "guest"]


def login(role: Role) -> None:
    print(role)
```
